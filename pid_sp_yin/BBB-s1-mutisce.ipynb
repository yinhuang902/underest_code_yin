```python
import numpy as np
import pyomo.environ as pyo
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D  # noqa: F401

# 注意：如果在 Jupyter Notebook 中运行，请取消下一行的注释
# %matplotlib widget


# ============================================================
# 0. 全局场景容器
# ============================================================
SCENARIOS = []


# ============================================================
# 1. 多场景真实目标 f(x,y) 以及 Pyomo 版
# ============================================================
def eval_single_gaussian(x, y, scen):
    """
    f_s(x,y) = - depth * exp(-||[x,y]-center||^2 / width^2)
               + outer_bias_strength * ||[x,y]-bias_center||^2
               + offset
    """
    cx, cy   = scen["center"]
    depth    = scen["depth"]
    width    = scen["width"]
    bs       = scen["outer_bias_strength"]
    bcx, bcy = scen["outer_bias_center"]
    offset   = scen["offset"]

    well_term = - depth * np.exp(-(((x-cx)**2 + (y-cy)**2)/(width**2)))
    bias_term = bs * ((x-bcx)**2 + (y-bcy)**2)
    return well_term + bias_term + offset


def f_true_numpy(xy):
    """
    期望目标: E[f(x,y)] = sum_s p_s * f_s(x,y)
    """
    x = float(xy[0])
    y = float(xy[1])
    total = 0.0
    for scen in SCENARIOS:
        ps = scen["p"]
        total += ps * eval_single_gaussian(x, y, scen)
    return total


def f_true_expr(m):
    """
    Pyomo 里同样的期望目标
    """
    expr = 0
    for scen in SCENARIOS:
        ps = float(scen["p"])
        cx, cy   = float(scen["center"][0]), float(scen["center"][1])
        depth    = float(scen["depth"])
        width    = float(scen["width"])
        bs       = float(scen["outer_bias_strength"])
        bcx, bcy = float(scen["outer_bias_center"][0]), float(scen["outer_bias_center"][1])
        offset   = float(scen["offset"])

        term_well = -depth * pyo.exp(-(((m.x - cx)**2 + (m.y - cy)**2)/(width**2)))
        term_bias = bs * ((m.x - bcx)**2 + (m.y - bcy)**2)
        expr += ps * (term_well + term_bias + offset)
    return expr


# ============================================================
# 2. 构建一阶段模型，读取变量的定义域
# ============================================================
def build_model():
    m = pyo.ConcreteModel()
    m.x = pyo.Var(bounds=(0.0, 10.0))
    m.y = pyo.Var(bounds=(0.0, 10.0))
    fsv = [m.x, m.y]  # first-stage vars
    return m, fsv


# ============================================================
# 3. 用一阶段变量的上下界自动生成角点
# ============================================================
def generate_corners(fsv):
    lows  = []
    highs = []
    for v in fsv:
        lo = v.lb
        hi = v.ub
        if lo is None or hi is None:
            raise ValueError("bounds missing on first-stage var")
        lows.append(float(lo))
        highs.append(float(hi))

    lows  = np.array(lows, dtype=float)
    highs = np.array(highs, dtype=float)

    d = len(fsv)
    nodes = []
    for mask in range(1 << d):
        point = []
        for i in range(d):
            point.append(highs[i] if ((mask >> i) & 1) else lows[i])
        nodes.append(np.array(point, dtype=float))
    return nodes


# ============================================================
# 4. 评估期望目标值
# ============================================================
def evaluate_objective_at(node):
    return f_true_numpy(node)


# ============================================================
# 5. 梯度：对每个场景求偏导，再加权
# ============================================================
def grad_single_gaussian(x, y, scen):
    cx, cy   = scen["center"]
    depth    = scen["depth"]
    width    = scen["width"]
    bs       = scen["outer_bias_strength"]
    bcx, bcy = scen["outer_bias_center"]

    dx = x - cx
    dy = y - cy
    r2 = dx**2 + dy**2

    # d/dx of -depth * exp(-r2/width^2)
    coef = depth * np.exp(-(r2/(width**2))) * (2.0/(width**2))
    dfx_well = coef * dx
    dfy_well = coef * dy

    # d/dx of bias_term = bs * ( (x-bcx)^2 + (y-bcy)^2 )
    dfx_bias = bs * 2.0*(x-bcx)
    dfy_bias = bs * 2.0*(y-bcy)

    dfx = dfx_well + dfx_bias
    dfy = dfy_well + dfy_bias
    return dfx, dfy


def compute_grad_at_point(xy):
    x = float(xy[0])
    y = float(xy[1])
    gx = 0.0
    gy = 0.0
    for scen in SCENARIOS:
        ps = scen["p"]
        dfx_s, dfy_s = grad_single_gaussian(x, y, scen)
        gx += ps * dfx_s
        gy += ps * dfy_s
    return np.array([gx, gy], dtype=float)


# ============================================================
# 6. 针对单个"片段"（基于某个采样点 i）求 ms_i
#    As_i(x,y) = f_i + g_i · ([x,y]-[x_i,y_i])
#    解: min_x [ f(x,y) - As_i(x,y) ]
#    得到:
#        ms_i   = 最小值 (通常负)
#        x_ms_i = 达到最小值的位置
# ============================================================
def solve_ms_for_slice(node_i, f_i, g_i, solver='ipopt', tee=False):
    xi, yi = float(node_i[0]), float(node_i[1])
    gx, gy = float(g_i[0]), float(g_i[1])

    m = pyo.ConcreteModel()
    m.x = pyo.Var(bounds=(0.0, 10.0), initialize=5.0)  # 添加初始值
    m.y = pyo.Var(bounds=(0.0, 10.0), initialize=5.0)  # 添加初始值

    # As_i(x,y) = f_i + gx*(x-xi) + gy*(y-yi)
    m.As_i = f_i + gx*(m.x - xi) + gy*(m.y - yi)

    # 目标: min ( f_true - As_i )
    m.obj = pyo.Objective(expr=f_true_expr(m) - m.As_i,
                          sense=pyo.minimize)

    opt = pyo.SolverFactory(solver)
    if solver == "gurobi":
        opt.options["NonConvex"] = 2
    res = opt.solve(m, tee=tee)

    x_star = np.array([pyo.value(m.x), pyo.value(m.y)], dtype=float)
    ms_val = float(pyo.value(m.obj))               # ms_i
    f_at_x = f_true_numpy(x_star)                  # f(x_ms_i)
    As_at_x = float(f_i + gx*(x_star[0]-xi) + gy*(x_star[1]-yi))

    return {
        "slice_index": None,     # 迭代外面会填
        "base_node": node_i,     # 这个片段来自哪个已有点
        "f_base": f_i,           # f(x_i)
        "grad": g_i,             # ∇f(x_i)
        "x_ms": x_star,          # 差距最负的位置
        "ms": ms_val,            # ms_i
        "f_xms": f_at_x,         # f(x_ms_i)
        "As_xms": As_at_x,       # As_i(x_ms_i)
        "status": str(res.solver.status),
        "term": str(res.solver.termination_condition),
    }


# ============================================================
# 7. A(x), A_shift(x) 以及节点上的 A(x_j)
# ============================================================
def A_on_grid(nodes, values, grads, xs, ys):
    """
    A(x,y) = max_i [ f_i + g_i·((x,y)-(x_i,y_i)) ]
    """
    X, Y = np.meshgrid(xs, ys)
    A = np.full_like(X, -np.inf, dtype=float)  # 使用 -np.inf 更安全
    for (pt, val, g) in zip(nodes, values, grads):
        xi, yi = pt
        gx, gy = g
        plane = val + gx*(X - xi) + gy*(Y - yi)
        A = np.maximum(A, plane)
    return X, Y, A


def Ashift_on_grid(nodes, values, grads, ms_list, xs, ys):
    """
    A_shift(x,y) = max_i [ As_i(x,y) + ms_i ]
                 = max_i [ f_i + g_i·((x,y)-(x_i,y_i)) + ms_i ]
    这是"加了ms之后的piecewise function"，我们将用它画成橙色曲面
    """
    X, Y = np.meshgrid(xs, ys)
    A_s = np.full_like(X, -np.inf, dtype=float)  # 使用 -np.inf 更安全
    for (pt, val, g, ms_i) in zip(nodes, values, grads, ms_list):
        xi, yi = pt
        gx, gy = g
        plane_shifted = val + gx*(X - xi) + gy*(Y - yi) + ms_i
        A_s = np.maximum(A_s, plane_shifted)
    return X, Y, A_s


def A_at_points(nodes, values, grads, eval_pts):
    """
    对每个给定点 (x_j,y_j) 计算:
    A(x_j) = max_i As_i(x_j)
    其中 As_i(x) = f_i + g_i · ((x,y)-(x_i,y_i))
    返回 list
    """
    A_list = []
    for xj, yj in eval_pts:
        best = -np.inf
        for (pt, val, g) in zip(nodes, values, grads):
            xi, yi = pt
            gx, gy = g
            As_val = val + gx*(xj - xi) + gy*(yj - yi)
            best = max(best, As_val)
        A_list.append(best)
    return A_list


# ============================================================
# 8. 缺失的 global_piecewise_step 函数
# ============================================================
def global_piecewise_step(nodes, values, grads, solver='ipopt', tee=False):
    """
    全局分段线性逼近的一步迭代
    1. 对每个现有节点i，求解 ms_i = min_x [f(x) - As_i(x)]
    2. 计算 LB = min_j [A(x_j) + ms_iter]，其中 ms_iter = min_i ms_i
    3. 计算 UB = min_j f(x_j) （当前最好的真实目标值）
    """
    # 计算每个片段的 ms
    per_slice = []
    for i, (node, val, grad) in enumerate(zip(nodes, values, grads)):
        result = solve_ms_for_slice(node, val, grad, solver=solver, tee=tee)
        result["slice_index"] = i
        per_slice.append(result)
    
    # 找出最负的 ms
    ms_list = [r["ms"] for r in per_slice]
    ms_iter = min(ms_list)
    best_idx = np.argmin(ms_list)
    
    # 计算 A(x_j) 在所有节点上的值
    A_vals = A_at_points(nodes, values, grads, nodes)
    
    # LB = min_j [A(x_j) + ms_iter]
    LB_iter = min(A_vals) + ms_iter
    
    # UB = min_j f(x_j)
    UB_iter = min(values)
    
    return per_slice, best_idx, LB_iter, UB_iter, ms_iter, ms_list


# ============================================================
# 9. 缺失的 pick_new_node_with_min_dist 函数
# ============================================================
def pick_new_node_with_min_dist(per_slice, existing_nodes, min_dist=1e-4):
    """
    从候选点中选择一个新点，确保它与现有点的距离大于 min_dist
    返回选中的记录和其排名
    """
    # 按 ms 从小到大排序
    sorted_slices = sorted(per_slice, key=lambda r: r["ms"])
    
    for rank, rec in enumerate(sorted_slices, start=1):
        candidate = rec["x_ms"]
        
        # 检查与所有现有点的距离
        too_close = False
        for node in existing_nodes:
            dist = np.linalg.norm(candidate - node)
            if dist < min_dist:
                too_close = True
                break
        
        if not too_close:
            return rec, rank
    
    return None, None  # 所有候选点都太近


# ============================================================
# 10. 3D 可视化
# ============================================================
def plot_iteration_global(iter_id, nodes, best_node_now, new_node, 
                         values, grads, ms_list, 
                         grid_n=60, elev=15, azim=220):
    fig = plt.figure(figsize=(10,8))
    ax = fig.add_subplot(111, projection='3d')

    nodes_arr = np.array(nodes)
    xs = np.linspace(0.0, 10.0, grid_n)
    ys = np.linspace(0.0, 10.0, grid_n)
    
    # 真实目标函数（深蓝色）
    X, Y = np.meshgrid(xs, ys)
    Z_true = np.zeros_like(X)
    for i in range(X.shape[0]):
        for j in range(X.shape[1]):
            Z_true[i,j] = f_true_numpy([X[i,j], Y[i,j]])

    ax.plot_surface(X, Y, Z_true,
                   cmap='Blues',
                   alpha=0.3,
                   linewidth=0,
                   antialiased=True)

    # A_shift 橙色曲面
    Xas, Yas, Zas = Ashift_on_grid(nodes, values, grads, ms_list, xs, ys)
    ax.plot_surface(Xas, Yas, Zas,
                   cmap='Oranges',
                   alpha=0.5,
                   linewidth=0,
                   antialiased=True)

    # 橙色线：从 A_shift 到真实曲面的垂直连线
    step = max(1, grid_n//15)
    for i in range(0, X.shape[0], step):
        for j in range(0, X.shape[1], step):
            x_pt = X[i, j]
            y_pt = Y[i, j]
            z_true = Z_true[i, j]
            z_as = Zas[i, j]
            if not np.isinf(z_as):  # 避免画无穷大的线
                ax.plot([x_pt, x_pt], [y_pt, y_pt], [z_as, z_true],
                       color='orange',
                       linewidth=1.0,
                       alpha=0.8)

    # 黑点：已采样节点
    node_z = np.array([f_true_numpy(p) for p in nodes_arr])
    ax.scatter(nodes_arr[:,0], nodes_arr[:,1], node_z,
               s=40, c='k', depthshade=False, label='Sampled nodes')

    # 当前最好点 (绿色)
    if best_node_now is not None:
        ax.scatter([best_node_now[0]], [best_node_now[1]],
                   [f_true_numpy(best_node_now)],
                   color='g', s=80, marker='o', label='Current min node')

    # 新的候选点 (蓝色)
    if new_node is not None:
        ax.scatter([new_node[0]], [new_node[1]],
                   [f_true_numpy(new_node)],
                   color='b', s=80, marker='o', label='Next node')

    ax.set_title(f"Global PW Iter {iter_id}")
    ax.set_xlabel("x")
    ax.set_ylabel("y")
    ax.set_zlabel("E[f(x,y)]")
    ax.legend(loc='upper left')
    ax.view_init(elev=elev, azim=azim)
    plt.tight_layout()
    plt.show()


# ============================================================
# 11. 主循环：全局 piecewise 策略
# ============================================================
def run_case_global(init_nodes,
                    max_nodes=10,
                    solver='ipopt',
                    tee=False,
                    plot_grid_n=60,
                    record_history=True,
                    make_3d_plots=True,
                    min_dist=1e-4,
                    print_tables=True):

    # 初始化
    nodes  = [np.array(p, dtype=float) for p in init_nodes]
    values = [evaluate_objective_at(p) for p in nodes]
    grads  = [compute_grad_at_point(p) for p in nodes]

    LB_hist, UB_hist, ms_hist, node_count = [], [], [], []

    it = 0
    while len(nodes) < max_nodes:

        # (1) 计算 per_slice / LB_iter / UB_iter / ms_iter / ms_list
        per_slice, best_idx, LB_iter, UB_iter, ms_iter, ms_list = global_piecewise_step(
            nodes, values, grads,
            solver=solver,
            tee=tee
        )

        # (2) 记录历史
        if record_history:
            LB_hist.append(LB_iter)
            UB_hist.append(UB_iter)
            ms_hist.append(ms_iter)
            node_count.append(len(nodes))

        # (3) 从 per_slice 里选新点（按 ms 排序，跳过太近）
        chosen_rec, chosen_rank = pick_new_node_with_min_dist(
            per_slice,
            nodes,
            min_dist=min_dist
        )
        if chosen_rec is None:
            print("All candidate x_ms are too close to existing nodes; stop.")
            break

        new_node = np.array(chosen_rec["x_ms"], dtype=float)
        used_ms  = chosen_rec["ms"]

        # (4) 打印诊断信息
        if print_tables:
            print(f"\n===== Iteration {it} (Global PW) =====")
            print(f"UB_iter (best true so far)               = {UB_iter:.6f}")
            print(f"LB_iter (min_j A(x_j) + ms_iter)         = {LB_iter:.6f}")
            print(f"Gap (UB_iter - LB_iter)                  = {UB_iter - LB_iter:.6f}")
            print(f"ms_iter (most negative ms_i overall)     = {ms_iter:.6e}")
            print(f"Chosen candidate rank {chosen_rank} with ms={used_ms:.6e}")
            print("\nPer-slice diagnostic (sorted by ms):")
            print(" idx | f_i=f(base)   ms_i        x_ms              f_xms")
            per_slice_sorted = sorted(per_slice, key=lambda r: r["ms"])
            for rank_print, r in enumerate(per_slice_sorted, start=1):
                mark = "<-- used" if rank_print == chosen_rank else ""
                print(f"{r['slice_index']:4d} | {r['f_base']:.6f}   {r['ms']:.3e}   "
                      f"({r['x_ms'][0]:.4f},{r['x_ms'][1]:.4f})   {r['f_xms']:.6f} {mark}")

        # (5) 画图（黑点 + 橙色加ms的piecewise + 橙色连线）
        if make_3d_plots and len(nodes[0]) == 2:
            idx_best_node = int(np.argmin(values))
            best_node_now = nodes[idx_best_node]
            plot_iteration_global(
                it,
                nodes,
                best_node_now,
                new_node,
                values,
                grads,
                ms_list,
                grid_n=plot_grid_n
            )

        # (6) 把新点加入样本集
        nodes.append(new_node)
        values.append(evaluate_objective_at(new_node))
        grads.append(compute_grad_at_point(new_node))

        it += 1

    return {
        "nodes": np.array(nodes),
        "values": np.array(values),
        "LB_hist": LB_hist,
        "UB_hist": UB_hist,
        "ms_hist": ms_hist,
        "node_count": node_count
    }


# ============================================================
# 12. 收敛曲线
# ============================================================
def plot_convergence_global(hist, prec=4):
    x  = hist["node_count"]
    LB = hist["LB_hist"]
    UB = hist["UB_hist"]
    ms = hist["ms_hist"]

    # LB / UB 曲线
    plt.figure(figsize=(6,4))
    plt.plot(x, LB, marker='o', label="LB_iter = min_j A(x_j)+ms_iter")
    plt.plot(x, UB, marker='s', label="UB_iter = min_j f(node_j)")
    plt.xlabel("Number of nodes")
    plt.ylabel("Bound value")
    plt.title("Global LB and UB vs Number of Nodes")
    plt.legend()
    plt.grid(True)
    plt.tight_layout()

    # ms 曲线
    plt.figure(figsize=(6,4))
    plt.plot(x, ms, marker='o')
    plt.xlabel("Number of nodes")
    plt.ylabel("ms_iter (most negative)")
    plt.title("ms vs Number of Nodes (Global piecewise)")
    plt.grid(True)
    plt.tight_layout()
    plt.show()

    # 打印表
    print("\n==== Summary Table (Global PW) ====")
    fmt = f"{{:.{prec}f}}"
    header = f"{'Nodes':>8s} | {'ms_iter':>12s} | {'LB_iter':>12s} | {'UB_iter':>12s}"
    print(header)
    print("-" * len(header))
    for n, m_val, lb, ub in zip(x, ms, LB, UB):
        print(f"{n:8d} | {fmt.format(m_val):>12s} | {fmt.format(lb):>12s} | {fmt.format(ub):>12s}")
    print("-" * len(header))
    print(f"{'Total':>8s} | {len(x):>12d}")


# ============================================================
# 13. 随机场景生成
# ============================================================
def random_scenarios(num_scenarios,
                     domain_low=0.0, domain_high=10.0,
                     depth_range=(0.8, 2.0),
                     width_range=(0.8, 2.0),
                     bias_strength_range=(0.02, 0.08),
                     bias_center=(5.0,5.0),
                     offset=0.0,
                     seed=None):
    rng = np.random.default_rng(seed)
    scens = []
    for _ in range(num_scenarios):
        cx = rng.uniform(domain_low, domain_high)
        cy = rng.uniform(domain_low, domain_high)
        depth = rng.uniform(*depth_range)
        width = rng.uniform(*width_range)
        bs = rng.uniform(*bias_strength_range)
        scens.append({
            "center": np.array([cx, cy], dtype=float),
            "depth": float(depth),
            "width": float(width),
            "outer_bias_strength": float(bs),
            "outer_bias_center": np.array(bias_center, dtype=float),
            "offset": float(offset),
        })
    # 均匀权重 p_s
    pval = 1.0/len(scens)
    for s in scens:
        s["p"] = pval
    return scens


# ============================================================
# 14. main
# ============================================================
if __name__ == "__main__":
    # 1) 生成场景
    SCENARIOS = random_scenarios(
        num_scenarios=3,
        seed=42
    )

    # 2) 用变量 bounds 生成初始角点
    model_for_bounds, fsv = build_model()
    init_nodes = generate_corners(fsv)

    # 3) 运行主循环
    solver      = "ipopt"   # 或 "gurobi"
    max_nodes   = 80
    plot_3d     = True
    record_hist = True

    hist_global = run_case_global(
        init_nodes=init_nodes,
        max_nodes=max_nodes,
        solver=solver,
        tee=False,
        plot_grid_n=60,
        record_history=record_hist,
        make_3d_plots=plot_3d,
        min_dist=1e-4,
        print_tables=True
    )

    # 4) 收敛曲线
    plot_convergence_global(hist_global, prec=4)

    # 5) 总结
    nodes  = hist_global["nodes"]
    values = hist_global["values"]
    print("\n==== Done (Global PW) ====")
    print(f"Total nodes: {len(nodes)}")
    print(f"Best expected f among nodes: {np.min(values):.6f}")
```

```python
import numpy as np
import pyomo.environ as pyo
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D  # noqa: F401

# 如果在 Jupyter Notebook 并想拖动交互图，请打开下一行
%matplotlib widget


# ============================================================
# 0. 全局场景容器
# ============================================================
SCENARIOS = []


# ============================================================
# 1. 多场景真实目标 f(x,y) 以及 Pyomo 版
# ============================================================
def eval_single_gaussian(x, y, scen):
    """
    f_s(x,y) = - depth * exp(-||[x,y]-center||^2 / width^2)
               + outer_bias_strength * ||[x,y]-bias_center||^2
               + offset
    """
    cx, cy   = scen["center"]
    depth    = scen["depth"]
    width    = scen["width"]
    bs       = scen["outer_bias_strength"]
    bcx, bcy = scen["outer_bias_center"]
    offset   = scen["offset"]

    well_term = - depth * np.exp(-(((x-cx)**2 + (y-cy)**2)/(width**2)))
    bias_term = bs * ((x-bcx)**2 + (y-bcy)**2)
    return well_term + bias_term + offset


def f_true_numpy(xy):
    """
    期望目标: E[f(x,y)] = sum_s p_s * f_s(x,y)
    """
    x = float(xy[0])
    y = float(xy[1])
    total = 0.0
    for scen in SCENARIOS:
        ps = scen["p"]
        total += ps * eval_single_gaussian(x, y, scen)
    return total


def f_true_expr(m):
    """
    Pyomo 里同样的期望目标
    """
    expr = 0
    for scen in SCENARIOS:
        ps = float(scen["p"])
        cx, cy   = float(scen["center"][0]), float(scen["center"][1])
        depth    = float(scen["depth"])
        width    = float(scen["width"])
        bs       = float(scen["outer_bias_strength"])
        bcx, bcy = float(scen["outer_bias_center"][0]), float(scen["outer_bias_center"][1])
        offset   = float(scen["offset"])

        term_well = -depth * pyo.exp(-(((m.x - cx)**2 + (m.y - cy)**2)/(width**2)))
        term_bias = bs * ((m.x - bcx)**2 + (m.y - bcy)**2)
        expr += ps * (term_well + term_bias + offset)
    return expr


# ============================================================
# 2. 构建一阶段模型，读取变量的定义域
# ============================================================
def build_model():
    m = pyo.ConcreteModel()
    m.x = pyo.Var(bounds=(0.0, 10.0))
    m.y = pyo.Var(bounds=(0.0, 10.0))
    fsv = [m.x, m.y]  # first-stage vars
    return m, fsv


# ============================================================
# 3. 用一阶段变量的上下界自动生成角点
# ============================================================
def generate_corners(fsv):
    lows  = []
    highs = []
    for v in fsv:
        lo = v.lb
        hi = v.ub
        if lo is None or hi is None:
            raise ValueError("bounds missing on first-stage var")
        lows.append(float(lo))
        highs.append(float(hi))

    lows  = np.array(lows, dtype=float)
    highs = np.array(highs, dtype=float)

    d = len(fsv)
    nodes = []
    for mask in range(1 << d):
        point = []
        for i in range(d):
            point.append(highs[i] if ((mask >> i) & 1) else lows[i])
        nodes.append(np.array(point, dtype=float))
    return nodes


# ============================================================
# 4. 评估期望目标值
# ============================================================
def evaluate_objective_at(node):
    return f_true_numpy(node)


# ============================================================
# 5. 梯度：对每个场景求偏导，再加权
# ============================================================
def grad_single_gaussian(x, y, scen):
    cx, cy   = scen["center"]
    depth    = scen["depth"]
    width    = scen["width"]
    bs       = scen["outer_bias_strength"]
    bcx, bcy = scen["outer_bias_center"]

    dx = x - cx
    dy = y - cy
    r2 = dx**2 + dy**2

    # d/dx of -depth * exp(-r2/width^2)
    coef = depth * np.exp(-(r2/(width**2))) * (2.0/(width**2))
    dfx_well = coef * dx
    dfy_well = coef * dy

    # d/dx of bias_term = bs * ( (x-bcx)^2 + (y-bcy)^2 )
    dfx_bias = bs * 2.0*(x-bcx)
    dfy_bias = bs * 2.0*(y-bcy)

    dfx = dfx_well + dfx_bias
    dfy = dfy_well + dfy_bias
    return dfx, dfy


def compute_grad_at_point(xy):
    x = float(xy[0])
    y = float(xy[1])
    gx = 0.0
    gy = 0.0
    for scen in SCENARIOS:
        ps = scen["p"]
        dfx_s, dfy_s = grad_single_gaussian(x, y, scen)
        gx += ps * dfx_s
        gy += ps * dfy_s
    return np.array([gx, gy], dtype=float)


# ============================================================
# 6. 针对单个“片段”（基于某个采样点 i）求 ms_i
#    As_i(x,y) = f_i + g_i · ([x,y]-[x_i,y_i])
#    解: min_x [ f(x,y) - As_i(x,y) ]
#    得到:
#        ms_i   = 最小值 (通常负)
#        x_ms_i = 达到最小值的位置
# ============================================================
def solve_ms_for_slice(node_i, f_i, g_i, solver='ipopt', tee=False):
    xi, yi = float(node_i[0]), float(node_i[1])
    gx, gy = float(g_i[0]), float(g_i[1])

    m = pyo.ConcreteModel()
    m.x = pyo.Var(bounds=(0.0, 10.0))
    m.y = pyo.Var(bounds=(0.0, 10.0))

    # As_i(x,y) = f_i + gx*(x-xi) + gy*(y-yi)
    m.As_i = f_i + gx*(m.x - xi) + gy*(m.y - yi)

    # 目标: min ( f_true - As_i )
    m.obj = pyo.Objective(expr=f_true_expr(m) - m.As_i,
                          sense=pyo.minimize)

    opt = pyo.SolverFactory(solver)
    if solver == "gurobi":
        opt.options["NonConvex"] = 2
    res = opt.solve(m, tee=tee)

    x_star = np.array([pyo.value(m.x), pyo.value(m.y)], dtype=float)
    ms_val = float(pyo.value(m.obj))               # ms_i
    f_at_x = f_true_numpy(x_star)                  # f(x_ms_i)
    As_at_x = float(f_i + gx*(x_star[0]-xi) + gy*(x_star[1]-yi))

    return {
        "slice_index": None,     # 迭代外面会填
        "base_node": node_i,     # 这个片段来自哪个已有点
        "f_base": f_i,           # f(x_i)
        "grad": g_i,             # ∇f(x_i)
        "x_ms": x_star,          # 差距最负的位置
        "ms": ms_val,            # ms_i
        "f_xms": f_at_x,         # f(x_ms_i)
        "As_xms": As_at_x,       # As_i(x_ms_i)
        "status": str(res.solver.status),
        "term": str(res.solver.termination_condition),
    }


# ============================================================
# 7. A(x), A_shift(x) 以及节点上的 A(x_j)
# ============================================================
def A_on_grid(nodes, values, grads, xs, ys):
    """
    A(x,y) = max_i [ f_i + g_i·((x,y)-(x_i,y_i)) ]
    """
    X, Y = np.meshgrid(xs, ys)
    A = np.full_like(X, -1e30, dtype=float)
    for (pt, val, g) in zip(nodes, values, grads):
        xi, yi = pt
        gx, gy = g
        plane = val + gx*(X - xi) + gy*(Y - yi)
        A = np.maximum(A, plane)
    return X, Y, A


def Ashift_on_grid(nodes, values, grads, ms_list, xs, ys):
    """
    A_shift(x,y) = max_i [ As_i(x,y) + ms_i ]
                 = max_i [ f_i + g_i·((x,y)-(x_i,y_i)) + ms_i ]
    这是“加了ms之后的piecewise function”，我们将用它画成橙色曲面
    """
    X, Y = np.meshgrid(xs, ys)
    A_s = np.full_like(X, -1e30, dtype=float)
    for (pt, val, g, ms_i) in zip(nodes, values, grads, ms_list):
        xi, yi = pt
        gx, gy = g
        plane_shifted = val + gx*(X - xi) + gy*(Y - yi) + ms_i
        A_s = np.maximum(A_s, plane_shifted)
    return X, Y, A_s


def A_at_points(nodes, values, grads, eval_pts):
    """
    对每个给定点 (x_j,y_j) 计算:
    A(x_j) = max_i As_i(x_j)
    其中 As_i(x) = f_i + g_i · ((x,y)-(x_i,y_i))
    返回 list
    """
    A_list = []
    for xj, yj in eval_pts:
        best = -1e30
        for (pt, val, g) in zip(nodes, values, grads):
            xi, yi = pt
            gx, gy = g
            est = val + gx*(xj - xi) + gy*(yj - yi)  # As_i(x_j)
            if est > best:
                best = est
        A_list.append(best)
    return A_list


# ============================================================
# 8. 聚合所有片段，生成本轮的指标
#    - per_slice: 每个切片的 ms_i 等
#    - UB_iter: min_j f(x_j)
#    - ms_iter: min_i ms_i
#    - LB_iter: 你最新定义的全局LB
#        LB_iter = min_j A(x_j) + ms_iter
#    - ms_list: 所有 ms_i（画橙色面要用）
# ============================================================
def global_piecewise_step(nodes, values, grads, solver='ipopt', tee=False):
    per_slice = []
    for i, (pt, val, g) in enumerate(zip(nodes, values, grads)):
        info = solve_ms_for_slice(pt, val, g, solver=solver, tee=tee)
        info["slice_index"] = i
        per_slice.append(info)

    # 当前真实上界: 目前采样点里的最小 f
    UB_iter = float(np.min(values))

    # 每个切片的 ms_i
    ms_list = [r["ms"] for r in per_slice]

    # 全局最坏 gap
    ms_iter = float(min(ms_list))

    # 计算 A(x_j) 在所有当前节点 x_j 处
    A_vals_at_nodes = A_at_points(
        nodes,
        values,
        grads,
        eval_pts=nodes  # 就是现有的插值点本身
    )

    # 你的定义的LB:
    # LB_iter = min_j A(x_j) + ms_iter
    LB_iter = float(min(A_vals_at_nodes) + ms_iter)

    # 最小 ms 的切片（用来提议下一个点）
    best_list = [r for r in per_slice if r["ms"] == ms_iter]
    best_idx = best_list[0]["slice_index"]

    return per_slice, best_idx, LB_iter, UB_iter, ms_iter, ms_list


# ============================================================
# 9. 选下一个新点（按 ms 从小到大，距离太近就跳）
# ============================================================
def pick_new_node_with_min_dist(per_slice, nodes, min_dist):
    ordered = sorted(per_slice, key=lambda r: r["ms"])  # ms越小越优先

    def far_enough(p):
        for q in nodes:
            if np.linalg.norm(p - q) < min_dist:
                return False
        return True

    for rank, rec in enumerate(ordered, start=1):
        cand = np.array(rec["x_ms"], dtype=float)
        if far_enough(cand):
            return rec, rank

    return None, None  # 没有可用的新点


# ============================================================
# 10. 每轮迭代的3D图
#     - 黑点：现有插值点在真实函数上
#     - 橙色面：A_shift(x) = max_i [As_i(x)+ms_i]
#     - 橙色连线：从橙色面取等 x / 等 y 截线
# ============================================================
def plot_iteration_global(iter_id,
                          nodes,
                          best_node_now,
                          new_node,
                          values,
                          grads,
                          ms_list,
                          grid_n=60,
                          elev=35,
                          azim=-135):

    nodes_arr = np.asarray(nodes, float)

    xs = np.linspace(0,10,grid_n)
    ys = np.linspace(0,10,grid_n)
    XX, YY = np.meshgrid(xs, ys)

    # 真实期望曲面
    ZZ_true = np.zeros_like(XX)
    for i in range(grid_n):
        for j in range(grid_n):
            ZZ_true[i,j] = f_true_numpy([XX[i,j], YY[i,j]])

    # 橙色面：加了 ms 的 piecewise 函数
    XXs, YYs, ZZs = Ashift_on_grid(nodes, values, grads, ms_list, xs, ys)

    fig = plt.figure(figsize=(8,6))
    ax = fig.add_subplot(111, projection='3d')

    # 真实曲面 (半透明蓝)
    ax.plot_surface(XX, YY, ZZ_true,
                    alpha=0.3,
                    linewidth=0,
                    antialiased=True,
                    color='C0')

    # 橙色面 (半透明橙)
    ax.plot_surface(XXs, YYs, ZZs,
                    alpha=0.6,
                    linewidth=0,
                    antialiased=True,
                    color='#ff7043')

    # 橙色连线（等y、等x截线）
    num_lines = 6
    y_idx_list = np.linspace(0, grid_n-1, num_lines, dtype=int)
    x_idx_list = np.linspace(0, grid_n-1, num_lines, dtype=int)

    # 等y切片: x方向曲线
    for yi in y_idx_list:
        ax.plot(xs,
                np.full_like(xs, ys[yi]),
                ZZs[yi, :],
                color='#ff7043',
                linewidth=1.5)

    # 等x切片: y方向曲线
    for xi in x_idx_list:
        ax.plot(np.full_like(ys, xs[xi]),
                ys,
                ZZs[:, xi],
                color='#ff7043',
                linewidth=1.0,
                alpha=0.8)

    # 黑点：已采样节点
    node_z = np.array([f_true_numpy(p) for p in nodes_arr])
    ax.scatter(nodes_arr[:,0], nodes_arr[:,1], node_z,
               s=40, c='k', depthshade=False, label='Sampled nodes')

    # 当前最好点 (绿色)
    if best_node_now is not None:
        ax.scatter([best_node_now[0]], [best_node_now[1]],
                   [f_true_numpy(best_node_now)],
                   color='g', s=80, marker='o', label='Current min node')

    # 新的候选点 (蓝色)
    if new_node is not None:
        ax.scatter([new_node[0]], [new_node[1]],
                   [f_true_numpy(new_node)],
                   color='b', s=80, marker='o', label='Next node')

    ax.set_title(f"Global PW Iter {iter_id}")
    ax.set_xlabel("x")
    ax.set_ylabel("y")
    ax.set_zlabel("E[f(x,y)]")
    ax.legend(loc='upper left')
    ax.view_init(elev=elev, azim=azim)
    plt.tight_layout()
    plt.show()


# ============================================================
# 11. 主循环：全局 piecewise 策略
# ============================================================
def run_case_global(init_nodes,
                    max_nodes=10,
                    solver='ipopt',
                    tee=False,
                    plot_grid_n=60,
                    record_history=True,
                    make_3d_plots=True,
                    min_dist=1e-4,
                    print_tables=True):

    # 初始化
    nodes  = [np.array(p, dtype=float) for p in init_nodes]
    values = [evaluate_objective_at(p) for p in nodes]
    grads  = [compute_grad_at_point(p) for p in nodes]

    LB_hist, UB_hist, ms_hist, node_count = [], [], [], []

    it = 0
    while len(nodes) < max_nodes:

        # (1) 计算 per_slice / LB_iter / UB_iter / ms_iter / ms_list
        per_slice, best_idx, LB_iter, UB_iter, ms_iter, ms_list = global_piecewise_step(
            nodes, values, grads,
            solver=solver,
            tee=tee
        )

        # (2) 记录历史
        if record_history:
            LB_hist.append(LB_iter)
            UB_hist.append(UB_iter)
            ms_hist.append(ms_iter)
            node_count.append(len(nodes))

        # (3) 从 per_slice 里选新点（按 ms 排序，跳过太近）
        chosen_rec, chosen_rank = pick_new_node_with_min_dist(
            per_slice,
            nodes,
            min_dist=min_dist
        )
        if chosen_rec is None:
            print("All candidate x_ms are too close to existing nodes; stop.")
            break

        new_node = np.array(chosen_rec["x_ms"], dtype=float)
        used_ms  = chosen_rec["ms"]

        # (4) 打印诊断信息
        if print_tables:
            print(f"\n===== Iteration {it} (Global PW) =====")
            print(f"UB_iter (best true so far)               = {UB_iter:.6f}")
            print(f"LB_iter (min_j A(x_j) + ms_iter)         = {LB_iter:.6f}")
            print(f"Gap (UB_iter - LB_iter)                  = {UB_iter - LB_iter:.6f}")
            print(f"ms_iter (most negative ms_i overall)     = {ms_iter:.6e}")
            print(f"Chosen candidate rank {chosen_rank} with ms={used_ms:.6e}")
            print("\nPer-slice diagnostic (sorted by ms):")
            print(" idx | f_i=f(base)   ms_i        x_ms              f_xms")
            per_slice_sorted = sorted(per_slice, key=lambda r: r["ms"])
            for rank_print, r in enumerate(per_slice_sorted, start=1):
                mark = "<-- used" if rank_print == chosen_rank else ""
                print(f"{r['slice_index']:4d} | {r['f_base']:.6f}   {r['ms']:.3e}   "
                      f"({r['x_ms'][0]:.4f},{r['x_ms'][1]:.4f})   {r['f_xms']:.6f} {mark}")

        # (5) 画图（黑点 + 橙色加ms的piecewise + 橙色连线）
        if make_3d_plots and len(nodes[0]) == 2:
            idx_best_node = int(np.argmin(values))
            best_node_now = nodes[idx_best_node]
            plot_iteration_global(
                it,
                nodes,
                best_node_now,
                new_node,
                values,
                grads,
                ms_list,
                grid_n=plot_grid_n
            )

        # (6) 把新点加入样本集
        nodes.append(new_node)
        values.append(evaluate_objective_at(new_node))
        grads.append(compute_grad_at_point(new_node))

        it += 1

    return {
        "nodes": np.array(nodes),
        "values": np.array(values),
        "LB_hist": LB_hist,
        "UB_hist": UB_hist,
        "ms_hist": ms_hist,
        "node_count": node_count
    }


# ============================================================
# 12. 收敛曲线
# ============================================================
def plot_convergence_global(hist, prec=4):
    x  = hist["node_count"]
    LB = hist["LB_hist"]
    UB = hist["UB_hist"]
    ms = hist["ms_hist"]

    # LB / UB 曲线
    plt.figure(figsize=(6,4))
    plt.plot(x, LB, marker='o', label="LB_iter = min_j A(x_j)+ms_iter")
    plt.plot(x, UB, marker='s', label="UB_iter = min_j f(node_j)")
    plt.xlabel("Number of nodes")
    plt.ylabel("Bound value")
    plt.title("Global LB and UB vs Number of Nodes")
    plt.legend()
    plt.grid(True)
    plt.tight_layout()

    # ms 曲线
    plt.figure(figsize=(6,4))
    plt.plot(x, ms, marker='o')
    plt.xlabel("Number of nodes")
    plt.ylabel("ms_iter (most negative)")
    plt.title("ms vs Number of Nodes (Global piecewise)")
    plt.grid(True)
    plt.tight_layout()
    plt.show()

    # 打印表
    print("\n==== Summary Table (Global PW) ====")
    fmt = f"{{:.{prec}f}}"
    header = f"{'Nodes':>8s} | {'ms_iter':>12s} | {'LB_iter':>12s} | {'UB_iter':>12s}"
    print(header)
    print("-" * len(header))
    for n, m_val, lb, ub in zip(x, ms, LB, UB):
        print(f"{n:8d} | {fmt.format(m_val):>12s} | {fmt.format(lb):>12s} | {fmt.format(ub):>12s}")
    print("-" * len(header))
    print(f"{'Total':>8s} | {len(x):>12d}")


# ============================================================
# 13. 随机场景生成
# ============================================================
def random_scenarios(num_scenarios,
                     domain_low=0.0, domain_high=10.0,
                     depth_range=(0.8, 2.0),
                     width_range=(0.8, 2.0),
                     bias_strength_range=(0.02, 0.08),
                     bias_center=(5.0,5.0),
                     offset=0.0,
                     seed=None):
    rng = np.random.default_rng(seed)
    scens = []
    for _ in range(num_scenarios):
        cx = rng.uniform(domain_low, domain_high)
        cy = rng.uniform(domain_low, domain_high)
        depth = rng.uniform(*depth_range)
        width = rng.uniform(*width_range)
        bs = rng.uniform(*bias_strength_range)
        scens.append({
            "center": np.array([cx, cy], dtype=float),
            "depth": float(depth),
            "width": float(width),
            "outer_bias_strength": float(bs),
            "outer_bias_center": np.array(bias_center, dtype=float),
            "offset": float(offset),
        })
    # 均匀权重 p_s
    pval = 1.0/len(scens)
    for s in scens:
        s["p"] = pval
    return scens


# ============================================================
# 14. main
# ============================================================
if __name__ == "__main__":
    # 1) 生成场景
    SCENARIOS = random_scenarios(
        num_scenarios=3,
        seed=42
    )

    # 2) 用变量 bounds 生成初始角点
    model_for_bounds, fsv = build_model()
    init_nodes = generate_corners(fsv)

    # 3) 运行主循环
    solver      = "ipopt"   # 或 "gurobi"
    max_nodes   = 40
    plot_3d     = True
    record_hist = True

    hist_global = run_case_global(
        init_nodes=init_nodes,
        max_nodes=max_nodes,
        solver=solver,
        tee=False,
        plot_grid_n=60,
        record_history=record_hist,
        make_3d_plots=plot_3d,
        min_dist=1e-4,
        print_tables=True
    )

    # 4) 收敛曲线
    plot_convergence_global(hist_global, prec=4)

    # 5) 总结
    nodes  = hist_global["nodes"]
    values = hist_global["values"]
    print("\n==== Done (Global PW) ====")
    print(f"Total nodes: {len(nodes)}")
    print(f"Best expected f among nodes: {np.min(values):.6f}")
```

```python
import numpy as np
from scipy.spatial import Delaunay
import pyomo.environ as pyo
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D  # noqa: F401
from mpl_toolkits.mplot3d.art3d import Poly3DCollection
# %matplotlib widget  

# ============================================================
# 0. 全局场景容器（会在 main 里填充）
# ============================================================
SCENARIOS = []


# ============================================================
# 1. 多场景目标函数建模
#    全局真实目标(期望) = sum_s p_s * f_s(x,y)
# ============================================================

def eval_single_gaussian(x, y, scen):
    """
    f_s(x,y) = - depth * exp(- ||(x,y)-center||^2 / width^2 )
               + outer_bias_strength * ||(x,y)-bias_center||^2
               + offset
    """
    cx, cy   = scen["center"]
    depth    = scen["depth"]
    width    = scen["width"]
    bs       = scen["outer_bias_strength"]
    bcx, bcy = scen["outer_bias_center"]
    offset   = scen["offset"]

    well_term = - depth * np.exp(-(((x-cx)**2 + (y-cy)**2)/(width**2)))
    bias_term = bs * ((x-bcx)**2 + (y-bcy)**2)
    return well_term + bias_term + offset


def f_true_numpy(xy):
    """
    E[f(x,y)] = sum_s p_s * f_s(x,y)
    """
    x = float(xy[0])
    y = float(xy[1])
    total = 0.0
    for scen in SCENARIOS:
        ps = scen["p"]
        total += ps * eval_single_gaussian(x, y, scen)
    return total


def f_true_expr(m):
    """
    Pyomo表达式版本的期望目标 E[f(x,y)]
    """
    expr = 0
    for scen in SCENARIOS:
        ps = float(scen["p"])
        cx, cy   = float(scen["center"][0]), float(scen["center"][1])
        depth    = float(scen["depth"])
        width    = float(scen["width"])
        bs       = float(scen["outer_bias_strength"])
        bcx, bcy = float(scen["outer_bias_center"][0]), float(scen["outer_bias_center"][1])
        offset   = float(scen["offset"])

        term_well = -depth * pyo.exp(-(((m.x - cx)**2 + (m.y - cy)**2)/(width**2)))
        term_bias = bs * ((m.x - bcx)**2 + (m.y - bcy)**2)
        expr += ps * (term_well + term_bias + offset)
    return expr


# ============================================================
# 2. 构建Pyomo模型，并返回fsv (first-stage vars list)
# ============================================================
def build_model():
    m = pyo.ConcreteModel()
    # 这里的bounds会定义我们的搜索域
    m.x = pyo.Var(bounds=(0.0, 10.0))
    m.y = pyo.Var(bounds=(0.0, 10.0))

    # fsv顺序 = 坐标顺序
    fsv = [m.x, m.y]
    return m, fsv


# ============================================================
# 3. 根据fsv自动生成角点 (2^d个corners)
#    读取每个first-stage变量的lb/ub，构造所有组合
# ============================================================
def generate_corners(fsv):
    lows  = []
    highs = []
    for v in fsv:
        lo = v.lb
        hi = v.ub
        if lo is None or hi is None:
            raise ValueError(
                f"Variable {v.name} is missing finite bounds: lb={lo}, ub={hi}"
            )
        if lo > hi:
            raise ValueError(
                f"Variable {v.name} has lb > ub: lb={lo}, ub={hi}"
            )
        lows.append(float(lo))
        highs.append(float(hi))

    lows  = np.array(lows, dtype=float)
    highs = np.array(highs, dtype=float)

    d = len(fsv)
    if d == 0:
        raise ValueError("No first-stage variables in fsv.")

    nodes = []
    for mask in range(1 << d):  # 遍历 0..2^d-1
        point = []
        for i in range(d):
            if (mask >> i) & 1:
                point.append(highs[i])
            else:
                point.append(lows[i])
        nodes.append(np.array(point, dtype=float))
    return nodes


# ============================================================
# 4. 评估角点的目标值(期望)
#    将来可以换成二阶段求解+加权，这里直接用f_true_numpy
# ============================================================
def evaluate_objective_at(node):
    return f_true_numpy(node)


# ============================================================
# 5. 在每个三角形里解 ms, LB, UB
#    注意所有值都是"期望目标"
# ============================================================
def minimize_on_each_triangle(nodes, values, solver='ipopt', tee=False):
    """
    nodes: list of R^2 points
    values: list of f_true_numpy(node) (期望值)
    返回:
      tri: Delaunay剖分
      results: 每个三角形的ms/LB/UB等信息
    """
    pts  = np.asarray(nodes, dtype=float)   # (N,2)
    vals = np.asarray(values, dtype=float)  # (N,)
    tri = Delaunay(pts)                     # tri.simplices: (K,3)

    # 我们用一个"基模型"的结构，然后在循环里clone，避免污染
    m_base, _ = build_model()
    opt = pyo.SolverFactory(solver)
    if solver == "gurobi":
        opt.options["NonConvex"] = 2

    results = []
    for k, simp in enumerate(tri.simplices):
        verts  = pts[simp]     # (3,2)
        fverts = vals[simp]    # (3,)

        # clone独立模型，避免累积约束冲突
        m = m_base.clone()

        # barycentric λ_j
        m.lam = pyo.Var(range(3), domain=pyo.NonNegativeReals)
        m.lam_sum = pyo.Constraint(expr=sum(m.lam[j] for j in range(3)) == 1.0)

        # x = Σ λ_j * v_jx,  y = Σ λ_j * v_jy
        m.x_link_x = pyo.Constraint(
            expr=m.x == sum(m.lam[j]*float(verts[j,0]) for j in range(3))
        )
        m.x_link_y = pyo.Constraint(
            expr=m.y == sum(m.lam[j]*float(verts[j,1]) for j in range(3))
        )

        # As = Σ λ_j * f(v_j)  (线性插值的代理值，对应期望值的线性插值)
        m.As = pyo.Var()
        m.As_con = pyo.Constraint(
            expr=m.As == sum(m.lam[j]*float(fverts[j]) for j in range(3))
        )

        # 目标: min [ E[f(x,y)] - As ]
        m.obj = pyo.Objective(expr=f_true_expr(m) - m.As, sense=pyo.minimize)

        res = opt.solve(m, tee=tee)

        x_ms = np.array([pyo.value(m.x), pyo.value(m.y)], dtype=float)
        ms_val = float(pyo.value(m.obj))       # 该三角形的 ms
        f_at_x_ms = f_true_numpy(x_ms)         # 期望函数在 x_ms 的真实值

        # LB = min As + ms = min f(v) + ms
        LB = float(np.min(fverts) + ms_val)

        # UB = min( min f(v), f(x_ms) )
        UB = float(min(np.min(fverts), f_at_x_ms))

        results.append({
            "simplex_index": int(k),
            "vertices_index": simp.tolist(),
            "verts": verts,
            "fverts": fverts,
            "x_ms": x_ms,
            "ms": ms_val,
            "LB": LB,
            "UB": UB,
            "f(x_ms)": f_at_x_ms,
            "status": str(res.solver.status),
            "term": str(res.solver.termination_condition),
        })

    return tri, results


# ============================================================
# 6. 全局LB的口径：v*是当前nodes中期望值最小的点
#    LB_iter = f(v*) + max_{Δ包含v*} ms_Δ
# ============================================================
def compute_lb_min_node_max_ms(nodes, values, per_tri):
    idx_min = int(np.argmin(values))
    ms_candidates = [r['ms'] for r in per_tri if idx_min in r['vertices_index']]
    if ms_candidates:
        return float(values[idx_min] + max(ms_candidates)), idx_min, float(max(ms_candidates))
    # fallback（几乎不会发生，除非退化三角形等）
    fallback = float(min(r['LB'] for r in per_tri))
    return fallback, idx_min, float('nan')


# ============================================================
# 7. 选点策略（和你原来的一样）
# ============================================================
def pick_candidate_by_ms_rank(per_tri, nodes, min_dist=1e-6, descending=False):
    """
    排序 per_tri: 默认 ms 从小到大 (descending=False)
    然后找一个 x_ms 距离现有节点都 >= min_dist 的
    """
    cands = sorted(per_tri, key=lambda r: r['ms'], reverse=descending)

    def ok(p):
        return all(np.linalg.norm(p - q) >= min_dist for q in nodes)

    for rank, r in enumerate(cands, start=1):
        p = np.array(r['x_ms'], float)
        if ok(p):
            return p, rank, float(r['ms'])

    return None, None, None


def print_per_triangle_table(results, active_mask):
    results = sorted(results, key=lambda r: r["simplex_index"])
    simplex_ids = [r["simplex_index"] for r in results]
    active_set = {sid for sid in simplex_ids if active_mask[sid]}

    header = ["row\\simp"] + [f"T{sid}{'*' if sid in active_set else ''}" for sid in simplex_ids]
    rows = [
        ["UB"] + [f"{r['UB']:.6f}" for r in results],
        ["LB"] + [f"{r['LB']:.6f}" for r in results],
        ["ms"] + [f"{r['ms']:.3e}" for r in results],
    ]
    table = [header] + rows

    ncols = len(header)
    colw = [0]*ncols
    for c in range(ncols):
        colw[c] = max(len(str(row[c])) for row in table) + 2

    RED = "\033[31m"
    RESET = "\033[0m"

    def colorize_if_active(col_idx: int, s: str) -> str:
        if col_idx == 0:
            return s
        sid = simplex_ids[col_idx-1]
        return f"{RED}{s}{RESET}" if sid in active_set else s

    header_line = "".join(
        str(header[c]).ljust(colw[c]) if c == 0
        else colorize_if_active(c, str(header[c]).ljust(colw[c]))
        for c in range(ncols)
    )

    print("\n== Per-triangle summary ==")
    print(header_line)
    print("-" * sum(colw))

    for r in rows:
        line = []
        for c in range(ncols):
            cell = str(r[c])
            padded = cell.ljust(colw[c]) if c == 0 else cell.rjust(colw[c])
            line.append(colorize_if_active(c, padded))
        print("".join(line))

    print("(红色列 = active；第1行=UB，第2行=LB，第3行=ms)\n")


# ============================================================
# 8. 每一轮的3D可视化 (期望曲面 + active单形屋顶)
# ============================================================
def plot_iteration(iter_id, nodes, tri, active_mask,
                   grid_n=60, elev=35, azim=-135,
                   min_node=None, new_node=None,
                   per_tri=None):
    nodes = np.asarray(nodes, float)

    COLOR_SURFACE = 'C0'           # 真实期望函数曲面
    COLOR_NODE_EDGES = '#444444'   # active 单形在真实曲面上的边
    COLOR_PIECE = '#ff7043'        # As+ms 屋顶
    LW_MAIN = 2.2

    # 画期望函数曲面
    # 注意：我们这里用的是 build_model 里的 bounds，先取[0,10]×[0,10]固定网格
    xs = np.linspace(0, 10, grid_n)
    ys = np.linspace(0, 10, grid_n)
    XX, YY = np.meshgrid(xs, ys)
    ZZ = np.zeros_like(XX)
    for i in range(XX.shape[0]):
        for j in range(XX.shape[1]):
            ZZ[i,j] = f_true_numpy([XX[i,j], YY[i,j]])

    fig = plt.figure(figsize=(8,6))
    ax = fig.add_subplot(111, projection='3d')

    ax.plot_surface(XX, YY, ZZ, alpha=0.35, linewidth=0,
                    antialiased=True, color=COLOR_SURFACE)

    # active单形的 As+ms 屋顶
    if per_tri is not None and len(per_tri) > 0 and tri is not None:
        faces = []
        for r in per_tri:
            sid = r["simplex_index"]
            if not active_mask.get(sid, False):
                continue
            verts_xy = r["verts"]     # (3,2)
            fverts   = r["fverts"]    # (3,)
            ms_val   = r["ms"]
            tri_xyz  = [
                (verts_xy[i,0], verts_xy[i,1], float(fverts[i] + ms_val))
                for i in range(3)
            ]
            faces.append(tri_xyz)

        if faces:
            coll = Poly3DCollection(
                faces,
                facecolors=[COLOR_PIECE],
                edgecolors='none',
                alpha=0.55
            )
            ax.add_collection3d(coll)

        # active单形屋顶的边线
        for r in per_tri:
            sid = r["simplex_index"]
            if not active_mask.get(sid, False):
                continue
            verts_xy = r["verts"]
            fverts   = r["fverts"]
            ms_val   = r["ms"]
            for i in range(3):
                a_xy = verts_xy[i]
                b_xy = verts_xy[(i+1) % 3]
                za = float(fverts[i] + ms_val)
                zb = float(fverts[(i+1) % 3] + ms_val)
                ax.plot([a_xy[0], b_xy[0]],
                        [a_xy[1], b_xy[1]],
                        [za, zb],
                        lw=LW_MAIN, c=COLOR_PIECE)

    # 所有节点 (黑点)
    node_z = np.array([f_true_numpy(p) for p in nodes])
    ax.scatter(nodes[:,0], nodes[:,1], node_z,
               s=30, c='k', depthshade=False)

    # active 单形在真实曲面上的边 + 质心
    if tri is not None:
        for k, simp in enumerate(tri.simplices):
            if not active_mask.get(k, False):
                continue
            verts_xy = tri.points[simp]
            for i in range(3):
                a, b = verts_xy[i], verts_xy[(i+1) % 3]
                ax.plot([a[0], b[0]], [a[1], b[1]],
                        [f_true_numpy(a), f_true_numpy(b)],
                        lw=LW_MAIN, c=COLOR_NODE_EDGES)

            centroid = np.mean(verts_xy, axis=0)
            ax.scatter([centroid[0]], [centroid[1]],
                       [f_true_numpy(centroid)],
                       marker='^', c=COLOR_NODE_EDGES,
                       s=70, depthshade=False)

    # 当前最优节点 (绿色)
    if min_node is not None:
        ax.scatter([min_node[0]], [min_node[1]],
                   [f_true_numpy(min_node)],
                   color='g', s=80, marker='o', label='Current min node')

    # 下一轮即将加入的新节点 (蓝色)
    if new_node is not None:
        ax.scatter([new_node[0]], [new_node[1]],
                   [f_true_numpy(new_node)],
                   color='b', s=80, marker='o', label='Next node')

    ax.set_title(f"Iteration {iter_id}")
    ax.set_xlabel("x")
    ax.set_ylabel("y")
    ax.set_zlabel("E[f(x,y)]")
    ax.legend(loc='upper left')
    ax.view_init(elev=elev, azim=azim)
    plt.tight_layout()
    plt.show()


# ============================================================
# 9. 收敛曲线 (LB/UB/ms vs node_count)
# ============================================================
def plot_convergence(hist, prec=4):
    x  = hist["node_count"]
    LB = hist["LB_hist"]
    UB = hist["UB_hist"]
    ms = hist["ms_hist"]

    # LB / UB
    plt.figure(figsize=(6,4))
    plt.plot(x, LB, marker='o', label="LB (lower bound)")
    plt.plot(x, UB, marker='s', label="UB (upper bound)")
    plt.xlabel("Number of nodes")
    plt.ylabel("Bound value")
    plt.title("Global LB and UB vs Number of Nodes")
    plt.legend()
    plt.grid(True)
    plt.tight_layout()

    # ms
    plt.figure(figsize=(6,4))
    plt.plot(x, ms, marker='o', color='C2')
    plt.xlabel("Number of nodes")
    plt.ylabel("Global ms (min over active simplices)")
    plt.title("ms vs Number of Nodes (Simplex method)")
    plt.grid(True)
    plt.tight_layout()
    plt.show()

    # 表格
    print("\n==== Summary Table ====")
    fmt = f"{{:.{prec}f}}"
    header = f"{'Nodes':>8s} | {'ms':>12s} | {'LB':>12s} | {'UB':>12s}"
    print(header)
    print("-" * len(header))
    for n, m_val, lb, ub in zip(x, ms, LB, UB):
        print(f"{n:8d} | {fmt.format(m_val):>12s} | {fmt.format(lb):>12s} | {fmt.format(ub):>12s}")
    print("-" * len(header))
    print(f"{'Total':>8s} | {len(x):>12d}")


# ============================================================
# 10. 主循环 run_case
#     和单场景改造后一样：不再自己造初始点
#     吃 init_nodes / init_values
# ============================================================
def run_case(init_nodes,
             init_values,
             max_nodes=10,
             solver='ipopt',
             tee=False,
             plot_grid_n=60,
             record_history=True,
             make_3d_plots=True,
             only_from_active=True,
             min_dist=1e-4,
             active_tol=1e-8,
             print_tables=True):
    """
    init_nodes: generate_corners(fsv) 的输出
    init_values: [ evaluate_objective_at(node) for node in init_nodes ]
    其他参数同以前
    """
    # 拷贝，避免原对象被原地修改
    nodes  = [np.array(p, dtype=float) for p in init_nodes]
    values = [float(v) for v in init_values]

    LB_hist, UB_hist, ms_hist, node_count = [], [], [], []

    it = 0
    while len(nodes) < max_nodes:
        tri, per_tri = minimize_on_each_triangle(
            nodes, values, solver=solver, tee=tee
        )

        # UB_global = 当前节点集中期望目标的最小值
        UB_global = float(np.min(values))

        # LB_iter = v*的值 + 其相邻单形 ms 的最大值
        LB_iter, idx_min_node, ms_chosen = compute_lb_min_node_max_ms(
            nodes, values, per_tri
        )

        # ms_iter = 所有单形里的最小 ms
        ms_iter = min(r['ms'] for r in per_tri)

        # 历史记录
        if record_history:
            LB_hist.append(LB_iter)
            UB_hist.append(UB_global)
            ms_hist.append(ms_iter)
            node_count.append(len(nodes))

        # active 单形
        active_mask = {
            r['simplex_index']: (r['LB'] <= UB_global + active_tol)
            for r in per_tri
        }
        active_tris = [r for r in per_tri if r['LB'] <= UB_global + active_tol]

        # 选择候选集合
        candidates = active_tris if (only_from_active and len(active_tris) > 0) else per_tri
        if only_from_active and len(active_tris) == 0:
            print(f"[warn] No active simplices this round (tol={active_tol:g}); "
                  f"fall back to all simplices.")

        # 选新节点：按 ms 最小优先，同时保持最小距离
        new_node, used_rank, used_ms = pick_candidate_by_ms_rank(
            candidates, nodes,
            min_dist=min_dist,
            descending=False
        )
        if new_node is None:
            print("All candidates too close; stop.")
            break

        # 输出本轮表格
        if print_tables:
            print(f"\n===== Iteration {it} =====")
            print(f"UB_global = {UB_global:.6f}")
            print(f"LB_iter   = {LB_iter:.6f}")
            print(f"ms_iter   = {ms_iter:.6e}")
            print_per_triangle_table(per_tri, active_mask)

        # 画3D (只有2D的一阶段变量我们才能画这个曲面图)
        if make_3d_plots and len(nodes[0]) == 2:
            plot_iteration(
                it, nodes, tri, active_mask,
                grid_n=plot_grid_n,
                min_node=nodes[idx_min_node],
                new_node=new_node,
                per_tri=per_tri
            )

        # 把新节点加入，如果和已有点太近就停止
        if any(np.linalg.norm(new_node - q) < min_dist for q in nodes):
            print("New node too close to an existing node; stop.")
            break

        nodes.append(new_node)
        values.append(evaluate_objective_at(new_node))

        it += 1

    return {
        "nodes": np.array(nodes),
        "values": np.array(values),
        "LB_hist": LB_hist,
        "UB_hist": UB_hist,
        "ms_hist": ms_hist,
        "node_count": node_count
    }


# ============================================================
# 11. 随机生成场景 (和你原来一样)
# ============================================================
def random_scenarios(num_scenarios,
                     domain_low=0.0, domain_high=10.0,
                     depth_range=(0.8, 2.0),
                     width_range=(0.8, 2.0),
                     bias_strength_range=(0.02, 0.08),
                     bias_center=(5.0,5.0),
                     offset=0.0,
                     seed=None):
    rng = np.random.default_rng(seed)
    scens = []
    for _ in range(num_scenarios):
        cx = rng.uniform(domain_low, domain_high)
        cy = rng.uniform(domain_low, domain_high)
        depth = rng.uniform(*depth_range)
        width = rng.uniform(*width_range)
        bs = rng.uniform(*bias_strength_range)
        scens.append({
            "center": np.array([cx, cy], dtype=float),
            "depth": float(depth),
            "width": float(width),
            "outer_bias_strength": float(bs),
            "outer_bias_center": np.array(bias_center, dtype=float),
            "offset": float(offset),
            # p 之后统一设置
        })
    # 均匀权重
    pval = 1.0/len(scens)
    for s in scens:
        s["p"] = pval
    return scens


# ============================================================
# 12. main：生成场景 -> 生成初始角点 -> 跑算法 -> 画收敛
# ============================================================
if __name__ == "__main__":
    # 1) 随机生成多场景
    SCENARIOS = random_scenarios(
        num_scenarios=3,
        seed=42  # 固定随机数种子以复现
    )

    # 2) 拿到一个模型 + fsv
    model_for_bounds, fsv = build_model()

    # 3) 用 fsv 的 bounds 自动生成所有角点
    init_nodes = generate_corners(fsv)

    # 4) 评估这些角点的期望目标值
    init_values = [evaluate_objective_at(p) for p in init_nodes]

    # 5) 运行主循环
    solver      = "ipopt"   # 也可以"gurobi"
    max_nodes   = 20
    plot_3d     = True
    record_hist = True

    hist_simp = run_case(
        init_nodes=init_nodes,
        init_values=init_values,
        max_nodes=max_nodes,
        solver=solver,
        tee=False,
        plot_grid_n=60,
        record_history=record_hist,
        make_3d_plots=plot_3d,
        only_from_active=True,
        min_dist=1e-4,
        active_tol=1e-8,
        print_tables=True
    )

    # 6) 收敛曲线
    plot_convergence(hist_simp, prec=4)

    # 7) 最后打印
    nodes  = hist_simp["nodes"]
    values = hist_simp["values"]
    print("\n==== Done ====")
    print(f"Total nodes: {len(nodes)}")
    print(f"Best expected f among nodes: {np.min(values):.6f}")
```
