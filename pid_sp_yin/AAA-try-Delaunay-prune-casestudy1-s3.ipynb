```python
# ==============================================
# Multi-Scenario Simplex Interpolation (2D)
# Gurobi + per-iteration plots + colored tables
# Save as: run_multi_simplex_gurobi.py
# ==============================================

import numpy as np
import pyomo.environ as pyo
from scipy.spatial import Delaunay
import matplotlib.pyplot as plt
from typing import List, Dict, Literal, Tuple
from mpl_toolkits.mplot3d import Axes3D  # noqa: F401
from mpl_toolkits.mplot3d.art3d import Poly3DCollection
%matplotlib widget

# ---------------- Config (global defaults) ----------------
AGG_DEFAULT     = 'mean'     # 'mean' | 'max' | 'cvar'
CVaR_ALPHA_DEF  = 0.9
ACTIVE_TOL_DEF  = 1e-8
MIN_DIST_DEF    = 1e-4
GRID_N_DEF      = 60
ELEV_DEF        = 35
AZIM_DEF        = -135
USE_GUROBI_OPT  = True

# ANSI color for terminal
RED   = "\033[31m"
RESET = "\033[0m"

AggT = Literal['mean','max','cvar']

# ----------------- Utilities: weighted stats -----------------
def weighted_quantile(vals: np.ndarray, w: np.ndarray, q: float) -> float:
    """离散加权分位数 VaR_q：最小的 v，使累计权重 >= q*W"""
    idx = np.argsort(vals)
    v = np.asarray(vals, float)[idx]
    ww = np.asarray(w, float)[idx]
    c = np.cumsum(ww)
    W = c[-1]
    j = int(np.searchsorted(c, q*W, side='left'))
    j = min(max(j, 0), len(v)-1)
    return float(v[j])

def weighted_cvar(vals: np.ndarray, w: np.ndarray, alpha: float) -> float:
    """
    加权CVaR_alpha（Expected Shortfall）：
    精确处理分位阈值所在样本的“部分权重”，避免用固定 eps。
    """
    idx = np.argsort(vals)
    v = np.asarray(vals, float)[idx]
    ww = np.asarray(w, float)[idx]
    c = np.cumsum(ww)
    W = c[-1]
    thr = alpha * W
    j = int(np.searchsorted(c, thr, side='left'))
    # 尾部总权重
    W_tail = W - (c[j-1] if j > 0 else 0.0)
    if W_tail <= 0:
        return float(v[-1])  # 退化保护
    # 分子：阈值位置 v[j] 只取 (c[j]-thr) 的部分权重，其余全取
    num = v[j] * (c[j] - thr)
    if j+1 < len(v):
        num += np.dot(v[j+1:], ww[j+1:])
    return float(num / W_tail)


# --------------- Scenario: quadratic model ------------------
def scenario_quadratic_numpy(theta: Dict, xy: np.ndarray) -> float:
    ax, ay = theta['ax'], theta['ay']
    cx, cy = theta['cx'], theta['cy']
    b0, bx, by = theta.get('b0', 0.0), theta.get('bx', 0.0), theta.get('by', 0.0)
    x, y = float(xy[0]), float(xy[1])
    return ax*(x-cx)**2 + ay*(y-cy)**2 + b0 + bx*x + by*y

def scenario_quadratic_expr(m: pyo.ConcreteModel, theta: Dict):
    ax, ay = theta['ax'], theta['ay']
    cx, cy = theta['cx'], theta['cy']
    b0, bx, by = theta.get('b0', 0.0), theta.get('bx', 0.0), theta.get('by', 0.0)
    return (ax*(m.x - float(cx))**2
            + ay*(m.y - float(cy))**2
            + float(b0) + float(bx)*m.x + float(by)*m.y)

def make_quadratic_scenarios(seed=42) -> Tuple[List[Dict], np.ndarray]:
    """
    ~48 scenarios (12 positions × 2 × 2 curvature), uniform weights.
    """
    rng = np.random.default_rng(seed)
    cx = rng.uniform(2.2, 3.8, size=12)
    cy = rng.uniform(4.2, 5.8, size=12)
    ax_choices = [1.7, 2.3]
    ay_choices = [2.7, 3.3]
    bx = rng.normal(0.0, 0.02, size=12)
    by = rng.normal(0.0, 0.02, size=12)

    thetas = []
    for i in range(12):
        for ax in ax_choices:
            for ay in ay_choices:
                thetas.append(dict(ax=ax, ay=ay,
                                   cx=float(cx[i]), cy=float(cy[i]),
                                   b0=0.0, bx=float(bx[i]), by=float(by[i])))
    w = np.ones(len(thetas), dtype=float)
    w /= w.sum()
    return thetas, w

# --------- Helper: scenario vertex values on a triangle ----------
def fverts_per_scenario(verts_xy: np.ndarray, scenarios: List[Dict]) -> np.ndarray:
    S = len(scenarios)
    out = np.zeros((S, 3), dtype=float)
    for s, th in enumerate(scenarios):
        for j in range(3):
            out[s, j] = scenario_quadratic_numpy(th, verts_xy[j])
    return out

# ------------- Per-triangle multi-scenario evaluation -------------
def minimize_on_each_triangle_multi(
    nodes: List[np.ndarray],
    scenarios: List[Dict],
    weights: np.ndarray,
    agg: AggT = AGG_DEFAULT,
    alpha: float = CVaR_ALPHA_DEF,
    solver: str = 'gurobi',
    tee: bool = False
):
    """
    For each triangle Δ:
      - Compute ms_s = min_{x∈Δ} [ f_s(x) - As_s(x) ] for each scenario s
      - Aggregate ms_Δ via mean / max / cvar(alpha)
      - Build LB_Δ = min_v (vertex f aggregated) + ms_Δ
      - Report UB_Δ = min( aggregated vertex value, aggregated f(x_ms_s) )
    Return tri, results(list of dict)
    """
    pts = np.asarray(nodes, float)
    tri = Delaunay(pts)

    # Pyomo model reused across triangles/scenarios
    m = pyo.ConcreteModel()
    m.x = pyo.Var(bounds=(0.0, 10.0))
    m.y = pyo.Var(bounds=(0.0, 10.0))
    m.lam = pyo.Var(range(3), domain=pyo.NonNegativeReals)
    m.lam_sum = pyo.Constraint(expr=sum(m.lam[j] for j in range(3)) == 1.0)
    m.x_link_x = pyo.Constraint(expr=m.x == 0.0)  # placeholder
    m.x_link_y = pyo.Constraint(expr=m.y == 0.0)  # placeholder
    m.As = pyo.Var()

    opt = pyo.SolverFactory(solver)
    if USE_GUROBI_OPT and solver.lower().startswith('gurobi'):
        opt.options.update({
            'OutputFlag': 0,
            'OptimalityTol': 1e-9,
            'BarConvTol': 1e-8,  # some Gurobi versions use different keys; harmless if ignored
            'NonConvex': 2, 
        })

    results = []
    S = len(scenarios)
    w = np.asarray(weights, float)
    w = w / w.sum()

    for k, simp in enumerate(tri.simplices):
        verts = pts[simp]  # (3,2)

        # bind x,y to barycentric coords of this triangle
        m.del_component(m.x_link_x); m.del_component(m.x_link_y)
        m.x_link_x = pyo.Constraint(expr=m.x == sum(m.lam[j]*float(verts[j,0]) for j in range(3)))
        m.x_link_y = pyo.Constraint(expr=m.y == sum(m.lam[j]*float(verts[j,1]) for j in range(3)))

        Fverts = fverts_per_scenario(verts, scenarios)  # (S,3)

        ms_s = np.zeros(S, dtype=float)
        x_ms_s = np.zeros((S,2), dtype=float)
        f_at_x_ms_s = np.zeros(S, dtype=float)

        # solve per-scenario ms
        for s, theta in enumerate(scenarios):
            # As_s = sum λ_j f_s(v_j)
            if hasattr(m, 'As_con'):
                m.del_component(m.As_con)
            m.As_con = pyo.Constraint(expr=m.As == sum(m.lam[j]*float(Fverts[s,j]) for j in range(3)))

            # objective: min f_s(x,y) - As_s
            if hasattr(m, 'obj'):
                m.del_component(m.obj)
            f_expr_s = scenario_quadratic_expr(m, theta)
            m.obj = pyo.Objective(expr=f_expr_s - m.As, sense=pyo.minimize)

            res = opt.solve(m, tee=tee)
            # (optional) check res.solver.termination_condition

            x_ms_s[s,:] = [pyo.value(m.x), pyo.value(m.y)]
            ms_s[s] = float(pyo.value(m.obj))
            f_at_x_ms_s[s] = float(scenario_quadratic_numpy(theta, x_ms_s[s,:]))

            # ---- 聚合 ms_Δ ----
            if agg == 'mean':
                ms_agg = float(np.sum(w*ms_s))
            elif agg == 'max':
                ms_agg = float(np.max(ms_s))
            elif agg == 'cvar':
                ms_agg = float(weighted_cvar(ms_s, w, alpha))
            else:
                raise ValueError("agg must be 'mean'|'max'|'cvar'")

            # 统计信息（保留原有）
            ms_mean = float(np.sum(w*ms_s))
            ms_max  = float(np.max(ms_s))
            ms_p90  = float(weighted_quantile(ms_s, w, 0.9))
            ms_std  = float(np.sqrt(np.sum(w*(ms_s - ms_mean)**2)))

            # ---- 新的：逐情景下界 LB_s，并按口径聚合成 LB_agg ----
            # 单形内线性插值 As 的最小值落在顶点 ⇒ min_x As_s(x) = min_j Fverts[s,j]
            LB_s = np.min(Fverts, axis=1) + ms_s  # 形状 (S,)
            if agg == 'mean':
                LB_agg = float(np.sum(w * LB_s))
            elif agg == 'max':
                LB_agg = float(np.max(LB_s))
            else:  # cvar
                LB_agg = float(weighted_cvar(LB_s, w, alpha))

            # ---- 报告用 UB_Δ：与原逻辑一致（不改）----
            # 顶点聚合值（仅用于 UB 的“顶点部分”对比；LB 不再使用 Gv）
            Gv = np.zeros(3, dtype=float)
            for j in range(3):
                vals = Fverts[:, j]
                if agg == 'mean':
                    Gv[j] = float(np.sum(w*vals))
                elif agg == 'max':
                    Gv[j] = float(np.max(vals))
                else:  # cvar
                    Gv[j] = float(weighted_cvar(vals, w, alpha))
            if agg == 'mean':
                f_x_ms_agg = float(np.sum(w*f_at_x_ms_s))
            elif agg == 'max':
                f_x_ms_agg = float(np.max(f_at_x_ms_s))
            else:
                f_x_ms_agg = float(weighted_cvar(f_at_x_ms_s, w, alpha))
            UB = float(min(np.min(Gv), f_x_ms_agg))

        # --------- (场景循环 for s 结束后) 聚合与统计 ---------
        # 1) 聚合 ms_Δ
        if agg == 'mean':
            ms_agg = float(np.sum(w*ms_s))
        elif agg == 'max':
            ms_agg = float(np.max(ms_s))
        elif agg == 'cvar':
            ms_agg = float(weighted_cvar(ms_s, w, alpha))
        else:
            raise ValueError("agg must be 'mean'|'max'|'cvar'")

        # 2) 统计信息
        ms_mean = float(np.sum(w*ms_s))
        ms_max  = float(np.max(ms_s))
        ms_p90  = float(weighted_quantile(ms_s, w, 0.9))
        ms_std  = float(np.sqrt(np.sum(w*(ms_s - ms_mean)**2)))

        # 3) 逐情景下界 LB_s，并按口径聚合得到 LB_agg（用于活跃判定/表格/LCB打分）
        #    线性插值 As 在三角形上最小值落在顶点 ⇒ min_x As_s(x) = min_j Fverts[s,j]
        LB_s = np.min(Fverts, axis=1) + ms_s  # 形状 (S,)
        if agg == 'mean':
            LB_agg = float(np.sum(w * LB_s))
        elif agg == 'max':
            LB_agg = float(np.max(LB_s))
        else:  # cvar
            LB_agg = float(weighted_cvar(LB_s, w, alpha))

        # 4) 报告用 UB_Δ：min( 顶点聚合值, f(x_ms_s) 的聚合 )
        #    顶点聚合（仅用于对比 UB；LB 已不依赖 Gv）
        Gv = np.zeros(3, dtype=float)
        for j in range(3):
            vals = Fverts[:, j]
            if agg == 'mean':
                Gv[j] = float(np.sum(w*vals))
            elif agg == 'max':
                Gv[j] = float(np.max(vals))
            else:
                Gv[j] = float(weighted_cvar(vals, w, alpha))
        if agg == 'mean':
            f_x_ms_agg = float(np.sum(w*f_at_x_ms_s))
        elif agg == 'max':
            f_x_ms_agg = float(np.max(f_at_x_ms_s))
        else:
            f_x_ms_agg = float(weighted_cvar(f_at_x_ms_s, w, alpha))
        UB = float(min(np.min(Gv), f_x_ms_agg))

        # 5) 代表点：取该三角形内 ms_s 最小的情景的 x_ms
        s_rep = int(np.argmin(ms_s))
        x_ms_rep = x_ms_s[s_rep, :]

        # 6) 记录结果
        results.append({
            "simplex_index": int(k),
            "vertices_index": simp.tolist(),
            "verts": verts,
            "Fverts": Fverts,         # (S,3)
            "ms_s": ms_s,             # (S,)
            "ms_agg": ms_agg,
            "ms_mean": ms_mean,
            "ms_max": ms_max,
            "ms_p90": ms_p90,
            "ms_std": ms_std,
            "LB": LB_agg,             # ✅ 用聚合后的真·下界
            "LB_s": LB_s,             # (可选) 逐情景下界
            "LB_agg": LB_agg,         # (可选) 同上
            "UB": UB,
            "x_ms": x_ms_rep,         # ✅ 代表点 = argmin(ms_s) 对应的解
            "x_ms_all": x_ms_s,       # 每个情景的 x_ms
            "f_x_ms_s": f_at_x_ms_s,  # 每个情景的 f(x_ms_s)
            "agg": agg,
        })



    return tri, results

# ----------------- Global UB & iter LB (aggregated) -----------------
def compute_global_UB_multi(
    nodes: List[np.ndarray],
    scenarios: List[Dict],
    weights: np.ndarray,
    agg: AggT = AGG_DEFAULT,
    alpha: float = CVaR_ALPHA_DEF
) -> float:
    w = np.asarray(weights, float); w = w / w.sum()
    vals = []
    for p in nodes:
        fs = np.array([scenario_quadratic_numpy(th, p) for th in scenarios], dtype=float)
        if agg == 'mean':
            agg_v = float(np.sum(w*fs))
        elif agg == 'max':
            agg_v = float(np.max(fs))
        else:
            agg_v = float(weighted_cvar(fs, w, alpha))
        vals.append(agg_v)
    return float(np.min(vals))

def compute_LB_iter_multi(
    nodes: List[np.ndarray],
    scenarios: List[Dict],
    weights: np.ndarray,
    per_tri: List[Dict],
    agg: AggT = AGG_DEFAULT,
    alpha: float = CVaR_ALPHA_DEF
):
    w = np.asarray(weights, float); w = w / w.sum()
    f_nodes_agg = []
    for p in nodes:
        fs = np.array([scenario_quadratic_numpy(th, p) for th in scenarios], dtype=float)
        if agg == 'mean':
            v = float(np.sum(w*fs))
        elif agg == 'max':
            v = float(np.max(fs))
        else:
            v = float(weighted_cvar(fs, w, alpha))
        f_nodes_agg.append(v)
    idx_min = int(np.argmin(f_nodes_agg))
    ms_cands = [r['ms_agg'] for r in per_tri if idx_min in r['vertices_index']]
    if ms_cands:
        return float(f_nodes_agg[idx_min] + max(ms_cands)), idx_min, float(max(ms_cands))
    return float(min(r['LB'] for r in per_tri)), idx_min, float('nan')

# -------------------- Candidate picking --------------------
def pick_candidate_by_ms_rank_multi(
    per_tri: List[Dict],
    nodes: List[np.ndarray],
    min_dist: float = 1e-6,
    kappa: float = 1.0
):
    """
    选单形：score = LB - kappa * ms_std （越小越优）
    单形内部代表点：该单形里 ms_s 最小情景的 x_ms（保持不变）
    """
    def tri_score(r):
        return r['LB'] - kappa * r['ms_std']   # 你的 LCB 打分

    cands = sorted(per_tri, key=tri_score)

    def ok(p):
        return all(np.linalg.norm(p - q) >= min_dist for q in nodes)

    for rank, r in enumerate(cands, start=1):
        p = np.array(r['x_ms'], float)  # ✅ 代表点仍用 ms_s 最小对应的 x_ms
        if ok(p):
            return p, rank, float(tri_score(r)), int(r['simplex_index'])
    return None, None, None, None



# ------------------- Pretty per-triangle table -------------------
def print_per_triangle_table(results, active_mask):
    results = sorted(results, key=lambda r: r["simplex_index"])
    simplex_ids = [r["simplex_index"] for r in results]
    active_set = {sid for sid in simplex_ids if active_mask.get(sid, False)}

    header = ["row\\simp"] + [f"T{sid}{'*' if sid in active_set else ''}" for sid in simplex_ids]
    rows = [
        ["UBΔ"]     + [f"{r['UB']:.6f}"      for r in results],
        ["LBΔ"]     + [f"{r['LB']:.6f}"      for r in results],
        ["ms_agg"]  + [f"{r['ms_agg']:.3e}"  for r in results],
        ["ms_mean"] + [f"{r['ms_mean']:.3e}" for r in results],
        ["ms_p90"]  + [f"{r['ms_p90']:.3e}"  for r in results],
        ["ms_max"]  + [f"{r['ms_max']:.3e}"  for r in results],
        ["ms_std"]  + [f"{r['ms_std']:.3e}"  for r in results],
    ]
    table = [header] + rows

    ncols = len(header)
    colw = [0]*ncols
    for c in range(ncols):
        colw[c] = max(len(str(row[c])) for row in table) + 2

    def colorize(col_idx: int, s: str) -> str:
        if col_idx == 0:  # row label
            return s
        sid = simplex_ids[col_idx-1]
        return f"{RED}{s}{RESET}" if sid in active_set else s

    header_line = "".join(str(header[c]).ljust(colw[c]) if c == 0
                          else colorize(c, str(header[c]).ljust(colw[c]))
                          for c in range(ncols))
    print("\n== Per-triangle summary ==")
    print(header_line)
    print("-" * sum(colw))
    for r in rows:
        line = []
        for c in range(ncols):
            cell = str(r[c])
            padded = cell.ljust(colw[c]) if c == 0 else cell.rjust(colw[c])
            line.append(colorize(c, padded))
        print("".join(line))
    print("(红色列 = active；行序: UBΔ, LBΔ, ms_agg, ms_mean, ms_p90, ms_max, ms_std)\n")

# ------------------- 3D plotting per iteration -------------------
def plot_iteration(
    iter_id: int,
    nodes: List[np.ndarray],
    tri: Delaunay,
    per_tri: List[Dict],
    active_mask: Dict[int, bool],
    grid_n: int = GRID_N_DEF,
    elev: int = ELEV_DEF,
    azim: int = AZIM_DEF,
    min_node=None,
    new_node=None
):
    nodes = np.asarray(nodes, float)
    # colors
    COLOR_SURFACE = 'C0'         # original aggregated surface
    COLOR_NODE_EDGES = '#444444' # active triangle edges on f-agg surface
    COLOR_PIECE = '#ff7043'      # As+ms face
    LW_MAIN = 2.0

    # aggregated surface over grid (using mean of scenarios for display)
    xs = np.linspace(0, 10, grid_n)
    ys = np.linspace(0, 10, grid_n)
    XX, YY = np.meshgrid(xs, ys)

    # use scenario mean for visualization (regardless of 'agg' just to show shape)
    # if you want strict consistency, you can recompute exact agg; mean is faster & smooth.
    def f_bar(x, y):
        return np.mean([scenario_quadratic_numpy(th, np.array([x,y])) for th in SCENARIOS], axis=0)

    ZZ = np.zeros_like(XX)
    for i in range(XX.shape[0]):
        for j in range(XX.shape[1]):
            ZZ[i,j] = f_bar(XX[i,j], YY[i,j])

    fig = plt.figure(figsize=(8,6))
    ax = fig.add_subplot(111, projection='3d')
    ax.plot_surface(XX, YY, ZZ, alpha=0.35, linewidth=0, antialiased=True, color=COLOR_SURFACE)

    # active As+ms faces
    faces = []
    for r in per_tri:
        sid = r["simplex_index"]
        if not active_mask.get(sid, False):
            continue
        verts_xy = r["verts"]     # (3,2)
        # Use As(face) + ms_agg => As at vertices equals aggregated vertex value we used? No:
        # For plotting, we lift triangle by (mean of scenario vertex values + ms_agg).
        # Use Gv_mean for a smoother face; exact agg face differs by agg choice (OK for vis).
        # Compute mean at vertices:
        Fverts = r["Fverts"]     # (S,3)
        vvals_mean = np.mean(Fverts, axis=0)  # (3,)
        ms = r["ms_agg"]
        tri_xyz  = [(verts_xy[i,0], verts_xy[i,1], float(vvals_mean[i] + ms)) for i in range(3)]
        faces.append(tri_xyz)

    if faces:
        coll = Poly3DCollection(faces, facecolors=[COLOR_PIECE], edgecolors='none', alpha=0.55)
        ax.add_collection3d(coll)

    # edges on aggregated f surface for active triangles
    def f_bar_np(xy): return f_bar(xy[0], xy[1])
    if tri is not None:
        for k, simp in enumerate(tri.simplices):
            if not active_mask.get(k, False):
                continue
            verts_xy = tri.points[simp]
            for i in range(3):
                a, b = verts_xy[i], verts_xy[(i+1) % 3]
                za = f_bar_np(a)
                zb = f_bar_np(b)
                ax.plot([a[0], b[0]], [a[1], b[1]], [za, zb], lw=LW_MAIN, c=COLOR_NODE_EDGES)

            # centroid marker (optional)
            centroid = np.mean(verts_xy, axis=0)
            ax.scatter([centroid[0]], [centroid[1]], [f_bar_np(centroid)],
                       marker='^', c=COLOR_NODE_EDGES, s=70, depthshade=False)

    # all nodes
    node_z = np.array([f_bar(p[0], p[1]) for p in nodes])
    ax.scatter(nodes[:,0], nodes[:,1], node_z, s=30, c='k', depthshade=False)

    # min node
    if min_node is not None:
        ax.scatter([min_node[0]], [min_node[1]], [f_bar(min_node[0], min_node[1])],
                   color='g', s=80, marker='o', label='Current min node')

    # next node
    if new_node is not None:
        ax.scatter([new_node[0]], [new_node[1]], [f_bar(new_node[0], new_node[1])],
                   color='b', s=80, marker='o', label='Next node')

    ax.set_title(f"Iteration {iter_id}")
    ax.set_xlabel("x"); ax.set_ylabel("y"); ax.set_zlabel("aggregated f (vis)")
    ax.legend(loc='upper left')
    ax.view_init(elev=elev, azim=azim)
    plt.tight_layout()
    plt.show()

# -------------------- Convergence plotting --------------------
def plot_convergence_multi(hist, prec=4):
    x  = hist["node_count"]
    LB = hist["LB_hist"]
    UB = hist["UB_hist"]
    ms = hist["ms_hist"]

    plt.figure(figsize=(6,4))
    plt.plot(x, LB, marker='o', label="LB (aggregated)")
    plt.plot(x, UB, marker='s', label="UB (aggregated)")
    plt.xlabel("Number of nodes"); plt.ylabel("Bound value")
    plt.title("Global LB and UB vs Number of Nodes (multi-scenario)")
    plt.legend(); plt.grid(True); plt.tight_layout()

    plt.figure(figsize=(6,4))
    plt.plot(x, ms, marker='o')
    plt.xlabel("Number of nodes"); plt.ylabel("Global ms (min over simplices; aggregated)")
    plt.title("ms vs Number of Nodes (multi-scenario)")
    plt.grid(True); plt.tight_layout(); plt.show()

    print("\n==== Summary Table (multi-scenario) ====")
    fmt = f"{{:.{prec}f}}"
    header = f"{'Nodes':>8s} | {'ms':>12s} | {'LB':>12s} | {'UB':>12s}"
    print(header); print("-"*len(header))
    for n, m, lb, ub in zip(x, ms, LB, UB):
        print(f"{n:8d} | {fmt.format(m):>12s} | {fmt.format(lb):>12s} | {fmt.format(ub):>12s}")
    print("-"*len(header))
    print(f"{'Total':>8s} | {len(x):>12d}")

# ------------------------ Main loop ------------------------
def run_case_multi(max_nodes=40,
                   solver='gurobi',
                   tee=False,
                   only_from_active=True,
                   min_dist=MIN_DIST_DEF,
                   active_tol=ACTIVE_TOL_DEF,
                   agg: AggT = AGG_DEFAULT,
                   alpha: float = CVaR_ALPHA_DEF,
                   seed=42,
                   record_history=True,
                   make_3d_plots=True):
    """
    Full iteration with per-iteration plots & tables.
    """
    global SCENARIOS  # used by plotting helper to build vis surface
    scenarios, weights = make_quadratic_scenarios(seed=seed)
    SCENARIOS = scenarios  # for visualization

    # initial 4 corners
    nodes = [np.array([0.0,0.0]), np.array([0.0,10.0]),
             np.array([10.0,0.0]), np.array([10.0,10.0])]

    LB_hist, UB_hist, ms_hist, node_count = [], [], [], []
    it = 0

    while len(nodes) < max_nodes:
        tri, per_tri = minimize_on_each_triangle_multi(
            nodes=nodes, scenarios=scenarios, weights=weights,
            agg=agg, alpha=alpha, solver=solver, tee=tee
        )
        UB_global = compute_global_UB_multi(
            nodes=nodes, scenarios=scenarios, weights=weights, agg=agg, alpha=alpha
        )
        LB_iter, idx_min_node, ms_used = compute_LB_iter_multi(
            nodes=nodes, scenarios=scenarios, weights=weights,
            per_tri=per_tri, agg=agg, alpha=alpha
        )
        ms_iter = min(r['ms_agg'] for r in per_tri)

        # active mask
        active_mask = { r['simplex_index']: (r['LB'] <= UB_global + active_tol) for r in per_tri }
        n_active = sum(1 for v in active_mask.values() if v)
        n_total  = len(per_tri)

        # -------------- iteration header log --------------
        print(f"\n=== Iter {it} | nodes={len(nodes)} | agg={agg} | alpha={alpha} ===")
        print(f"UB_global(agg) = {UB_global:.6f}")
        print(f"LB_iter(agg)   = {LB_iter:.6f}  [v* = node #{idx_min_node}, ms_used={ms_used:.3e}]")
        print(f"ms_iter(min ms_agg over simplices) = {ms_iter:.3e}")
        print(f"active simplices: {n_active}/{n_total}")

        # -------------- per-triangle table --------------
        print_per_triangle_table(per_tri, active_mask)
        # numerical sanity
        for r in per_tri:
            if r['LB'] > r['UB'] + 1e-7:
                print(f"{RED}[num_tol] T{r['simplex_index']}: LB({r['LB']:.6f}) > UB({r['UB']:.6f}){RESET}")

        # -------------- pick candidate --------------
        active_tris = [r for r in per_tri if active_mask[r['simplex_index']]]
        candidates = active_tris if (only_from_active and len(active_tris) > 0) else per_tri
        if only_from_active and len(active_tris) == 0:
            print(f"[warn] No active simplices this round (tol={active_tol:g}); fallback to all simplices.")

        KAPPA = 1.0  # 可调/可退火
        new_node, used_rank, used_score, used_tid = pick_candidate_by_ms_rank_multi(
            candidates, nodes, min_dist=min_dist, kappa=KAPPA
        )
        if new_node is None:
            print("All candidate x_ms are too close to existing nodes; stop.")
            break
        print(f"Selected triangle T{used_tid} (rank {used_rank}) with LCB-score = {used_score:.3e}  [score = LB - {KAPPA}*ms_std]")

        # -------------- visualization --------------
        if make_3d_plots:
            plot_iteration(
                iter_id=it, nodes=nodes, tri=tri, per_tri=per_tri, active_mask=active_mask,
                grid_n=GRID_N_DEF, elev=ELEV_DEF, azim=AZIM_DEF,
                min_node=nodes[idx_min_node], new_node=new_node
            )

        # -------------- record history --------------
        if record_history:
            LB_hist.append(LB_iter)
            UB_hist.append(UB_global)
            ms_hist.append(ms_iter)
            node_count.append(len(nodes))

        # -------------- add node --------------
        if any(np.linalg.norm(new_node - q) < min_dist for q in nodes):
            print("New node too close to an existing node; stop.")
            break
        nodes.append(new_node)

        it += 1
        if len(nodes) >= max_nodes:
            break

    return {
        "nodes": np.array(nodes),
        "LB_hist": LB_hist,
        "UB_hist": UB_hist,
        "ms_hist": ms_hist,
        "node_count": node_count,
        "agg": agg,
        "alpha": alpha,
        "scenarios": scenarios,
        "weights": weights,
    }

# ------------------------ Entrypoint ------------------------
if __name__ == "__main__":
    solver       = "gurobi"     # switch to "ipopt" if needed
    max_nodes    = 20
    agg          = "mean"       # 'mean' | 'max' | 'cvar'
    alpha        = 0.9
    only_active  = True
    min_dist     = 1e-4
    active_tol   = 1e-8
    seed         = 12
    make_plots   = True

    hist = run_case_multi(
        max_nodes=max_nodes,
        solver=solver,
        tee=False,
        only_from_active=only_active,
        min_dist=min_dist,
        active_tol=active_tol,
        agg=agg,
        alpha=alpha,
        seed=seed,
        record_history=True,
        make_3d_plots=make_plots
    )

    plot_convergence_multi(hist, prec=4)
    nodes = hist["nodes"]
    print("\n==== Done (multi-scenario, Gurobi) ====")
    print(f"Total nodes: {len(nodes)}")
    print(f"Best aggregated UB among nodes: {min(hist['UB_hist']) if hist['UB_hist'] else float('nan'):.6f}")
```
