```python
import numpy as np
from scipy.spatial import Delaunay
import pyomo.environ as pyo
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D  # noqa: F401
from mpl_toolkits.mplot3d.art3d import Poly3DCollection

# ============================================================
# 0. 全局场景容器（会在 main 里用 random_scenarios() 生成）
#    每个场景 s 是一个 dict，包含:
#    - "p": 概率
#    - "center": 高斯井中心
#    - "depth", "width": 井的深度和宽度
#    - "outer_bias_strength", "outer_bias_center", "offset": 外圈二次抬升
#    - "a","b": 资源耦合系数 (a_s, b_s)
# ============================================================
SCENARIOS = []


# ============================================================
# 1. 给定一阶段 y=(y1,y2)，场景 s 的"局部最优反应"
#    solve_second_stage_value(y, s)
#    这是真正的二阶段含义: 场景在 y 给的资源限制下选自己的 x_s
# ============================================================
def solve_second_stage_value(y_vec, scen, solver='ipopt'):
    """
    y_vec = [y1, y2] 是一阶段资源上限
    场景 scen 选择 (x1,x2) 来最小化它的高斯井代价:
        min f_s(x1,x2)
        s.t. x1 <= y1
             a_s * x1 + b_s * x2 <= y2
             x1,x2 >= 0
    返回该最优值 (float)
    """
    y1_val = float(y_vec[0])
    y2_val = float(y_vec[1])

    m = pyo.ConcreteModel()
    m.x1 = pyo.Var(bounds=(0.0, 10.0))
    m.x2 = pyo.Var(bounds=(0.0, 10.0))

    # 资源耦合约束
    a_s = scen["a"]
    b_s = scen["b"]
    m.c1 = pyo.Constraint(expr = m.x1 <= y1_val)
    m.c2 = pyo.Constraint(expr = a_s*m.x1 + b_s*m.x2 <= y2_val)

    # 高斯井 + 偏置
    cx, cy   = scen["center"]
    depth    = scen["depth"]
    width    = scen["width"]
    bs       = scen["outer_bias_strength"]
    bcx, bcy = scen["outer_bias_center"]
    offset   = scen["offset"]

    def obj_expr(m):
        well_term = - depth * pyo.exp(-(((m.x1-cx)**2 + (m.x2-cy)**2)/(width**2)))
        bias_term = bs * ((m.x1-bcx)**2 + (m.x2-bcy)**2)
        return well_term + bias_term + offset

    m.obj = pyo.Objective(rule=obj_expr, sense=pyo.minimize)

    opt = pyo.SolverFactory(solver)
    if solver == "gurobi":
        opt.options["NonConvex"] = 2

    res = opt.solve(m, tee=False)
    return float(pyo.value(m.obj))


# ============================================================
# 2. 给定 y=(y1,y2)，算期望目标 E_s[ Q_s(y) ]
#    也就是一阶段决策 y 的真实成本
# ============================================================
def f_true_numpy(y_vec, solver='ipopt'):
    """
    y_vec = [y1,y2]
    返回 sum_s p_s * Q_s(y)
    其中 Q_s(y) 是 solve_second_stage_value(y, scen)
    """
    total = 0.0
    for scen in SCENARIOS:
        ps = scen["p"]
        total += ps * solve_second_stage_value(y_vec, scen, solver=solver)
    return total


# ============================================================
# 3. 我们还需要一个"符号版"模型，用 Pyomo 表达期望目标
#    这个模型把所有场景同时摊在一层:
#    变量:
#       y1,y2 (一阶段)
#       x1[s], x2[s] (场景s的响应)
#    约束:
#       x1[s] <= y1
#       a_s*x1[s] + b_s*x2[s] <= y2
#       x1[s],x2[s] >=0
#    目标:
#       sum_s p_s * f_s(x1[s],x2[s])
#
#    注意: 这和上面的"二阶段优化后再加和"在数学上等价为
#          min_{y, x_s} sum_s p_s f_s(x_s) s.t. x_s feasible under y
# ============================================================
def build_master_model():
    """
    返回:
      m (ConcreteModel)
      fsv: list of first-stage variables [m.y1, m.y2]
    """
    m = pyo.ConcreteModel()

    # 一阶段变量 y1,y2 = 我们要在 (y1,y2) 空间里做Delaunay的那个东西
    m.y1 = pyo.Var(bounds=(0.0, 10.0))
    m.y2 = pyo.Var(bounds=(0.0, 10.0))

    # 索引场景
    m.scen_index = pyo.RangeSet(len(SCENARIOS)-1)

    # 每个场景的响应决策 x1[s], x2[s]
    m.x1 = pyo.Var(m.scen_index, bounds=(0.0, 10.0))
    m.x2 = pyo.Var(m.scen_index, bounds=(0.0, 10.0))

    # 资源耦合约束:
    # x1[s] <= y1
    def c1_rule(model, s):
        return model.x1[s] <= model.y1
    m.c1 = pyo.Constraint(m.scen_index, rule=c1_rule)

    # a_s * x1[s] + b_s * x2[s] <= y2
    def c2_rule(model, s):
        a_s = SCENARIOS[s]["a"]
        b_s = SCENARIOS[s]["b"]
        return a_s*model.x1[s] + b_s*model.x2[s] <= model.y2
    m.c2 = pyo.Constraint(m.scen_index, rule=c2_rule)

    # 目标: sum_s p_s f_s(x1[s],x2[s])
    def obj_rule(model):
        total = 0.0
        for s in model.scen_index:
            scen = SCENARIOS[s]
            p_s  = scen["p"]
            cx, cy = scen["center"]
            depth  = scen["depth"]
            width  = scen["width"]
            bs     = scen["outer_bias_strength"]
            bcx, bcy = scen["outer_bias_center"]
            offset = scen["offset"]

            x1s = model.x1[s]
            x2s = model.x2[s]

            well_term = - depth * pyo.exp(-(((x1s-cx)**2 + (x2s-cy)**2)/(width**2)))
            bias_term = bs * ((x1s-bcx)**2 + (x2s-bcy)**2)
            total += p_s * (well_term + bias_term + offset)
        return total

    m.obj = pyo.Objective(rule=obj_rule, sense=pyo.minimize)

    # fsv: first-stage (y1,y2)
    fsv = [m.y1, m.y2]
    return m, fsv


# ============================================================
# 4. 自动生成角点: 从fsv里读bounds，生成2^d个corners
# ============================================================
def generate_corners(fsv):
    lows  = []
    highs = []
    for v in fsv:
        lo = v.lb
        hi = v.ub
        if lo is None or hi is None:
            raise ValueError(f"Variable {v.name} missing finite bounds: {lo},{hi}")
        if lo > hi:
            raise ValueError(f"Variable {v.name} has lb>ub: {lo}>{hi}")
        lows.append(float(lo))
        highs.append(float(hi))

    lows  = np.array(lows, dtype=float)
    highs = np.array(highs, dtype=float)
    d = len(fsv)
    if d == 0:
        raise ValueError("No first-stage vars in fsv.")

    nodes = []
    for mask in range(1 << d):  # 0..2^d-1
        point = []
        for i in range(d):
            point.append(highs[i] if ((mask >> i) & 1) else lows[i])
        nodes.append(np.array(point, dtype=float))
    return nodes


# ============================================================
# 5. 给某个节点 y=(y1,y2) 评估真实目标 (期望值)
#    这里调用的是 "各场景单独二阶段优化后求和"
# ============================================================
def evaluate_objective_at(node, solver='ipopt'):
    return f_true_numpy(node, solver=solver)


# ============================================================
# 6. 在每个三角形里求 ms, LB, UB
#
# 现在我们不能再用旧的 f_true_expr(m) （已经拆掉了），
# 但我们有 build_master_model()，它就是“摊平版”的期望目标。
#
# 对某个三角形三顶点 v0,v1,v2 以及它们的真实值 f(vj)：
#   - 我们clone一份 master_model m
#   - 加 λ_j >=0, sum λ=1
#   - 约束 y1 = Σ λ_j * vj[0], y2 = Σ λ_j * vj[1]
#   - 定义 m.As = Σ λ_j * f(vj)
#   - 新目标: minimize ( master_obj_expr - m.As )
#
# 这里 master_obj_expr 就是 m.obj.expr (sum_s p_s f_s(x_s))
#
# 返回:
#   tri: Delaunay
#   results: list of dict with ms, LB, UB, etc.
# ============================================================
def minimize_on_each_triangle(nodes, values, solver='ipopt', tee=False):
    pts  = np.asarray(nodes, dtype=float)    # (N,2), each is y=(y1,y2)
    vals = np.asarray(values, dtype=float)   # (N,)
    tri = Delaunay(pts)                      # tri.simplices: (K,3)

    opt = pyo.SolverFactory(solver)
    if solver == "gurobi":
        opt.options["NonConvex"] = 2

    results = []
    for k, simp in enumerate(tri.simplices):
        verts  = pts[simp]     # (3,2)
        fverts = vals[simp]    # (3,)

        # 基于 master_model 复制一个模型
        m_base, _ = build_master_model()
        m = m_base.clone()

        # 让原本的目标先失效，等会儿建新目标 (ms目标)
        m.obj.deactivate()

        # λ_j 做重心坐标
        m.lam = pyo.Var(range(3), domain=pyo.NonNegativeReals)

        m.lam_sum = pyo.Constraint(
            expr=sum(m.lam[j] for j in range(3)) == 1.0
        )

        # y1 = Σ λ_j * v_j[0], y2 = Σ λ_j * v_j[1]
        m.link_y1 = pyo.Constraint(
            expr = m.y1 == sum(m.lam[j]*float(verts[j,0]) for j in range(3))
        )
        m.link_y2 = pyo.Constraint(
            expr = m.y2 == sum(m.lam[j]*float(verts[j,1]) for j in range(3))
        )

        # As = Σ λ_j * f(v_j)  (线性插值下估的那条"红折线")
        m.As = pyo.Var()
        m.As_con = pyo.Constraint(
            expr = m.As == sum(m.lam[j]*float(fverts[j]) for j in range(3))
        )

        # 真正的期望目标(摊平后): sum_s p_s f_s(x1[s],x2[s])
        master_expr = 0.0
        for s in m.scen_index:
            scen = SCENARIOS[s]
            p_s  = scen["p"]
            cx, cy = scen["center"]
            depth  = scen["depth"]
            width  = scen["width"]
            bs     = scen["outer_bias_strength"]
            bcx, bcy = scen["outer_bias_center"]
            offset = scen["offset"]

            x1s = m.x1[s]
            x2s = m.x2[s]

            well_term = - depth * pyo.exp(-(((x1s-cx)**2 + (x2s-cy)**2)/(width**2)))
            bias_term = bs * ((x1s-bcx)**2 + (x2s-bcy)**2)
            master_expr += p_s * (well_term + bias_term + offset)

        # 我们要最小化 "真实期望 - As"
        m.ms_obj = pyo.Objective(expr = master_expr - m.As, sense=pyo.minimize)

        res = opt.solve(m, tee=tee)

        # 求 ms 点 (y_ms = y1,y2 的取值；也可以看作 λ 找到的点）
        y_ms = np.array([pyo.value(m.y1), pyo.value(m.y2)], dtype=float)

        # ms 值
        ms_val = float(pyo.value(m.ms_obj))

        # 真实期望在这个 y_ms 上是多少？
        # 注意！这里用 solve_second_stage_value 场景逐个求真实Q_s(y_ms)
        # 而不是用 m 里的摊平解，因为 ms_obj只求了差值最小，不保证是最优x_s(y_ms)
        f_at_y_ms = f_true_numpy(y_ms, solver=solver)

        # LB = min f(v_j) + ms_val
        LB = float(np.min(fverts) + ms_val)

        # UB = min( min f(v_j), f_at_y_ms )
        UB = float(min(np.min(fverts), f_at_y_ms))

        results.append({
            "simplex_index": int(k),
            "vertices_index": simp.tolist(),
            "verts": verts,
            "fverts": fverts,
            "y_ms": y_ms,
            "ms": ms_val,
            "LB": LB,
            "UB": UB,
            "f(y_ms)": f_at_y_ms,
            "status": str(res.solver.status),
            "term": str(res.solver.termination_condition),
        })

    return tri, results


# ============================================================
# 7. 全局LB的口径：v*是当前nodes中期望值最小的点
#    LB_iter = f(v*) + max_{Δ包含v*} ms_Δ
# ============================================================
def compute_lb_min_node_max_ms(nodes, values, per_tri):
    idx_min = int(np.argmin(values))
    ms_candidates = [r['ms'] for r in per_tri if idx_min in r['vertices_index']]
    if ms_candidates:
        return float(values[idx_min] + max(ms_candidates)), idx_min, float(max(ms_candidates))
    # fallback
    fallback = float(min(r['LB'] for r in per_tri))
    return fallback, idx_min, float('nan')


# ============================================================
# 8. 选点策略：按 ms 最小优先，同时保持最小距离
# ============================================================
def pick_candidate_by_ms_rank(per_tri, nodes, min_dist=1e-6, descending=False):
    cands = sorted(per_tri, key=lambda r: r['ms'], reverse=descending)

    def ok(p):
        return all(np.linalg.norm(p - q) >= min_dist for q in nodes)

    for rank, r in enumerate(cands, start=1):
        p = np.array(r['y_ms'], float)
        if ok(p):
            return p, rank, float(r['ms'])
    return None, None, None


def print_per_triangle_table(results, active_mask):
    results = sorted(results, key=lambda r: r["simplex_index"])
    simplex_ids = [r["simplex_index"] for r in results]
    active_set = {sid for sid in simplex_ids if active_mask[sid]}

    header = ["row\\simp"] + [f"T{sid}{'*' if sid in active_set else ''}" for sid in simplex_ids]
    rows = [
        ["UB"] + [f"{r['UB']:.6f}" for r in results],
        ["LB"] + [f"{r['LB']:.6f}" for r in results],
        ["ms"] + [f"{r['ms']:.3e}" for r in results],
    ]
    table = [header] + rows

    ncols = len(header)
    colw = [0]*ncols
    for c in range(ncols):
        colw[c] = max(len(str(row[c])) for row in table) + 2

    RED = "\033[31m"
    RESET = "\033[0m"

    def colorize_if_active(col_idx: int, s: str) -> str:
        if col_idx == 0:
            return s
        sid = simplex_ids[col_idx-1]
        return f"{RED}{s}{RESET}" if sid in active_set else s

    header_line = "".join(
        str(header[c]).ljust(colw[c]) if c == 0
        else colorize_if_active(c, str(header[c]).ljust(colw[c]))
        for c in range(ncols)
    )

    print("\n== Per-triangle summary ==")
    print(header_line)
    print("-" * sum(colw))

    for r in rows:
        line = []
        for c in range(ncols):
            cell = str(r[c])
            padded = cell.ljust(colw[c]) if c == 0 else cell.rjust(colw[c])
            line.append(colorize_if_active(c, padded))
        print("".join(line))

    print("(红色列 = active；第1行=UB，第2行=LB，第3行=ms)\n")


# ============================================================
# 9. 3D可视化 (y1,y2 -> 期望成本曲面)
#    注意现在f_true_numpy()是"有资源耦合二阶段"的真实期望成本
# ============================================================
def plot_iteration(iter_id, nodes, tri, active_mask,
                   grid_n=40, elev=35, azim=-135,
                   min_node=None, new_node=None,
                   per_tri=None,
                   solver='ipopt'):
    nodes = np.asarray(nodes, float)

    COLOR_SURFACE = 'C0'           # 真实期望函数曲面
    COLOR_NODE_EDGES = '#444444'   # active 单形在真实曲面上的边
    COLOR_PIECE = '#ff7043'        # As+ms 屋顶
    LW_MAIN = 2.2

    # 网格画真实函数曲面
    xs = np.linspace(0, 10, grid_n)
    ys = np.linspace(0, 10, grid_n)
    XX, YY = np.meshgrid(xs, ys)
    ZZ = np.zeros_like(XX)
    for i in range(XX.shape[0]):
        for j in range(XX.shape[1]):
            ZZ[i,j] = f_true_numpy([XX[i,j], YY[i,j]], solver=solver)

    fig = plt.figure(figsize=(8,6))
    ax = fig.add_subplot(111, projection='3d')

    ax.plot_surface(XX, YY, ZZ, alpha=0.35, linewidth=0,
                    antialiased=True, color=COLOR_SURFACE)

    # active单形的 As+ms 屋顶
    if per_tri is not None and len(per_tri) > 0 and tri is not None:
        faces = []
        for r in per_tri:
            sid = r["simplex_index"]
            if not active_mask.get(sid, False):
                continue
            verts_xy = r["verts"]     # (3,2)
            fverts   = r["fverts"]    # (3,)
            ms_val   = r["ms"]
            tri_xyz  = [
                (verts_xy[i,0], verts_xy[i,1], float(fverts[i] + ms_val))
                for i in range(3)
            ]
            faces.append(tri_xyz)

        if faces:
            coll = Poly3DCollection(
                faces,
                facecolors=[COLOR_PIECE],
                edgecolors='none',
                alpha=0.55
            )
            ax.add_collection3d(coll)

        # active单形屋顶的边线
        for r in per_tri:
            sid = r["simplex_index"]
            if not active_mask.get(sid, False):
                continue
            verts_xy = r["verts"]
            fverts   = r["fverts"]
            ms_val   = r["ms"]
            for i in range(3):
                a_xy = verts_xy[i]
                b_xy = verts_xy[(i+1) % 3]
                za = float(fverts[i] + ms_val)
                zb = float(fverts[(i+1) % 3] + ms_val)
                ax.plot([a_xy[0], b_xy[0]], [a_xy[1], b_xy[1]],
                        [za, zb],
                        lw=LW_MAIN, c=COLOR_PIECE)

    # 所有采样节点 (黑点)
    node_z = np.array([f_true_numpy(p, solver=solver) for p in nodes])
    ax.scatter(nodes[:,0], nodes[:,1], node_z,
               s=30, c='k', depthshade=False)

    # active 单形在真实曲面上的边 + 质心
    if tri is not None:
        for k, simp in enumerate(tri.simplices):
            if not active_mask.get(k, False):
                continue
            verts_xy = tri.points[simp]
            for i in range(3):
                a, b = verts_xy[i], verts_xy[(i+1) % 3]
                ax.plot([a[0], b[0]], [a[1], b[1]],
                        [f_true_numpy(a, solver=solver),
                         f_true_numpy(b, solver=solver)],
                        lw=LW_MAIN, c=COLOR_NODE_EDGES)

            centroid = np.mean(verts_xy, axis=0)
            ax.scatter([centroid[0]], [centroid[1]],
                       [f_true_numpy(centroid, solver=solver)],
                       marker='^', c=COLOR_NODE_EDGES,
                       s=70, depthshade=False)

    # 当前最优节点 (绿色)
    if min_node is not None:
        ax.scatter([min_node[0]], [min_node[1]],
                   [f_true_numpy(min_node, solver=solver)],
                   color='g', s=80, marker='o', label='Current best node')

    # 下一轮待加入的新节点 (蓝色)
    if new_node is not None:
        ax.scatter([new_node[0]], [new_node[1]],
                   [f_true_numpy(new_node, solver=solver)],
                   color='b', s=80, marker='o', label='Next node')

    ax.set_title(f"Iteration {iter_id}")
    ax.set_xlabel("y1")
    ax.set_ylabel("y2")
    ax.set_zlabel("Expected cost with resource coupling")
    ax.legend(loc='upper left')
    ax.view_init(elev=elev, azim=azim)
    plt.tight_layout()
    plt.show()


# ============================================================
# 10. 收敛曲线 (LB/UB/ms vs node_count)
# ============================================================
def plot_convergence(hist, prec=4):
    x  = hist["node_count"]
    LB = hist["LB_hist"]
    UB = hist["UB_hist"]
    ms = hist["ms_hist"]

    # LB / UB
    plt.figure(figsize=(6,4))
    plt.plot(x, LB, marker='o', label="LB (lower bound)")
    plt.plot(x, UB, marker='s', label="UB (upper bound)")
    plt.xlabel("Number of nodes")
    plt.ylabel("Bound value")
    plt.title("Global LB and UB vs Number of Nodes")
    plt.legend()
    plt.grid(True)
    plt.tight_layout()

    # ms
    plt.figure(figsize=(6,4))
    plt.plot(x, ms, marker='o', color='C2')
    plt.xlabel("Number of nodes")
    plt.ylabel("Global ms (min over active simplices)")
    plt.title("ms vs Number of Nodes (Simplex method)")
    plt.grid(True)
    plt.tight_layout()
    plt.show()

    # 表格
    print("\n==== Summary Table ====")
    fmt = f"{{:.{prec}f}}"
    header = f"{'Nodes':>8s} | {'ms':>12s} | {'LB':>12s} | {'UB':>12s}"
    print(header)
    print("-" * len(header))
    for n, m_val, lb, ub in zip(x, ms, LB, UB):
        print(f"{n:8d} | {fmt.format(m_val):>12s} | {fmt.format(lb):>12s} | {fmt.format(ub):>12s}")
    print("-" * len(header))
    print(f"{'Total':>8s} | {len(x):>12d}")


# ============================================================
# 11. 主循环 run_case
# ============================================================
def run_case(init_nodes,
             init_values,
             max_nodes=10,
             solver='ipopt',
             tee=False,
             plot_grid_n=40,
             record_history=True,
             make_3d_plots=True,
             only_from_active=True,
             min_dist=1e-4,
             active_tol=1e-8,
             print_tables=True):
    """
    init_nodes: generate_corners(fsv) 的输出
    init_values: [evaluate_objective_at(node) ...]
    """
    nodes  = [np.array(p, dtype=float) for p in init_nodes]
    values = [float(v) for v in init_values]

    LB_hist, UB_hist, ms_hist, node_count = [], [], [], []

    it = 0
    while len(nodes) < max_nodes:
        tri, per_tri = minimize_on_each_triangle(
            nodes, values, solver=solver, tee=tee
        )

        # UB_global = 当前节点集中期望目标的最小值
        UB_global = float(np.min(values))

        # LB_iter = v*的值 + 其相邻单形 ms 的最大值
        LB_iter, idx_min_node, ms_chosen = compute_lb_min_node_max_ms(
            nodes, values, per_tri
        )

        # ms_iter = 所有单形里的最小 ms
        ms_iter = min(r['ms'] for r in per_tri)

        # 记录
        if record_history:
            LB_hist.append(LB_iter)
            UB_hist.append(UB_global)
            ms_hist.append(ms_iter)
            node_count.append(len(nodes))

        # active 单形
        active_mask = {
            r['simplex_index']: (r['LB'] <= UB_global + active_tol)
            for r in per_tri
        }
        active_tris = [r for r in per_tri if r['LB'] <= UB_global + active_tol]

        # 候选集合
        candidates = active_tris if (only_from_active and len(active_tris) > 0) else per_tri
        if only_from_active and len(active_tris) == 0:
            print(f"[warn] No active simplices (tol={active_tol:g}); fall back to all simplices.")

        # 找下一采样点（按 ms 最小）
        new_node, used_rank, used_ms = pick_candidate_by_ms_rank(
            candidates, nodes,
            min_dist=min_dist,
            descending=False
        )
        if new_node is None:
            print("All candidates too close; stop.")
            break

        # 打印
        if print_tables:
            print(f"\n===== Iteration {it} =====")
            print(f"UB_global = {UB_global:.6f}")
            print(f"LB_iter   = {LB_iter:.6f}")
            print(f"ms_iter   = {ms_iter:.6e}")
            print_per_triangle_table(per_tri, active_mask)

        # 可视化 (仅2D y空间时画3D曲面)
        if make_3d_plots and len(nodes[0]) == 2:
            plot_iteration(
                it, nodes, tri, active_mask,
                grid_n=plot_grid_n,
                min_node=nodes[idx_min_node],
                new_node=new_node,
                per_tri=per_tri,
                solver=solver
            )

        # 把新节点加入
        if any(np.linalg.norm(new_node - q) < min_dist for q in nodes):
            print("New node too close to an existing node; stop.")
            break

        nodes.append(new_node)
        values.append(evaluate_objective_at(new_node, solver=solver))

        it += 1

    return {
        "nodes": np.array(nodes),
        "values": np.array(values),
        "LB_hist": LB_hist,
        "UB_hist": UB_hist,
        "ms_hist": ms_hist,
        "node_count": node_count
    }


# ============================================================
# 12. 随机生成场景
#     加入 a,b 系数，用来控制资源约束:
#       x1_s <= y1
#       a_s * x1_s + b_s * x2_s <= y2
# ============================================================
def random_scenarios(num_scenarios,
                     domain_low=0.0, domain_high=10.0,
                     depth_range=(0.8, 2.0),
                     width_range=(0.8, 2.0),
                     bias_strength_range=(0.02, 0.08),
                     bias_center=(5.0,5.0),
                     offset=0.0,
                     ab_range=(0.5, 2.0),
                     seed=None):
    rng = np.random.default_rng(seed)
    scens = []
    for _ in range(num_scenarios):
        cx = rng.uniform(domain_low, domain_high)
        cy = rng.uniform(domain_low, domain_high)
        depth = rng.uniform(*depth_range)
        width = rng.uniform(*width_range)
        bs = rng.uniform(*bias_strength_range)
        a  = rng.uniform(*ab_range)
        b  = rng.uniform(*ab_range)
        scens.append({
            "center": np.array([cx, cy], dtype=float),
            "depth": float(depth),
            "width": float(width),
            "outer_bias_strength": float(bs),
            "outer_bias_center": np.array(bias_center, dtype=float),
            "offset": float(offset),
            "a": float(a),
            "b": float(b),
            # p 下面统一设
        })
    # 均匀权重
    pval = 1.0/len(scens)
    for s in scens:
        s["p"] = pval
    return scens


# ============================================================
# 13. main：生成场景 -> 初始角点 -> 跑算法 -> 画收敛
# ============================================================
if __name__ == "__main__":
    # 1) 随机生成多场景
    SCENARIOS = random_scenarios(
        num_scenarios=3,
        seed=42
    )

    # 2) 生成主模型（用于拿 bounds, 以及后续克隆）
    master_model_for_bounds, fsv = build_master_model()

    # 3) 用 fsv 的 bounds 自动生成所有角点 (初始采样点)
    init_nodes = generate_corners(fsv)

    # 4) 评估这些角点的期望目标值 (考虑资源耦合后的真实两阶段成本)
    init_values = [evaluate_objective_at(p, solver='ipopt') for p in init_nodes]

    # 5) 运行主循环
    solver      = "ipopt"   # 可换 "gurobi" (要 m.options["NonConvex"]=2)
    max_nodes   = 12
    plot_3d     = True
    record_hist = True

    hist_simp = run_case(
        init_nodes=init_nodes,
        init_values=init_values,
        max_nodes=max_nodes,
        solver=solver,
        tee=False,
        plot_grid_n=40,
        record_history=record_hist,
        make_3d_plots=plot_3d,
        only_from_active=True,
        min_dist=1e-4,
        active_tol=1e-8,
        print_tables=True
    )

    # 6) 收敛曲线
    plot_convergence(hist_simp, prec=4)

    # 7) 最后打印
    nodes  = hist_simp["nodes"]
    values = hist_simp["values"]
    print("\n==== Done ====")
    print(f"Total nodes: {len(nodes)}")
    print(f"Best expected f among nodes: {np.min(values):.6f}")
```

```python
import numpy as np
from scipy.spatial import Delaunay
import pyomo.environ as pyo
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D  # noqa: F401
from mpl_toolkits.mplot3d.art3d import Poly3DCollection
# %matplotlib widget  

# ============================================================
# 0. 全局场景容器（会在 main 里填充）
# ============================================================
SCENARIOS = []


# ============================================================
# 1. 多场景目标函数建模
#    全局真实目标(期望) = sum_s p_s * f_s(x,y)
# ============================================================

def eval_single_gaussian(x, y, scen):
    """
    f_s(x,y) = - depth * exp(- ||(x,y)-center||^2 / width^2 )
               + outer_bias_strength * ||(x,y)-bias_center||^2
               + offset
    """
    cx, cy   = scen["center"]
    depth    = scen["depth"]
    width    = scen["width"]
    bs       = scen["outer_bias_strength"]
    bcx, bcy = scen["outer_bias_center"]
    offset   = scen["offset"]

    well_term = - depth * np.exp(-(((x-cx)**2 + (y-cy)**2)/(width**2)))
    bias_term = bs * ((x-bcx)**2 + (y-bcy)**2)
    return well_term + bias_term + offset


def f_true_numpy(xy):
    """
    E[f(x,y)] = sum_s p_s * f_s(x,y)
    """
    x = float(xy[0])
    y = float(xy[1])
    total = 0.0
    for scen in SCENARIOS:
        ps = scen["p"]
        total += ps * eval_single_gaussian(x, y, scen)
    return total


def f_true_expr(m):
    """
    Pyomo表达式版本的期望目标 E[f(x,y)]
    """
    expr = 0
    for scen in SCENARIOS:
        ps = float(scen["p"])
        cx, cy   = float(scen["center"][0]), float(scen["center"][1])
        depth    = float(scen["depth"])
        width    = float(scen["width"])
        bs       = float(scen["outer_bias_strength"])
        bcx, bcy = float(scen["outer_bias_center"][0]), float(scen["outer_bias_center"][1])
        offset   = float(scen["offset"])

        term_well = -depth * pyo.exp(-(((m.x - cx)**2 + (m.y - cy)**2)/(width**2)))
        term_bias = bs * ((m.x - bcx)**2 + (m.y - bcy)**2)
        expr += ps * (term_well + term_bias + offset)
    return expr


# ============================================================
# 2. 构建Pyomo模型，并返回fsv (first-stage vars list)
# ============================================================
def build_model():
    m = pyo.ConcreteModel()
    # 这里的bounds会定义我们的搜索域
    m.x = pyo.Var(bounds=(0.0, 10.0))
    m.y = pyo.Var(bounds=(0.0, 10.0))

    # fsv顺序 = 坐标顺序
    fsv = [m.x, m.y]
    return m, fsv


# ============================================================
# 3. 根据fsv自动生成角点 (2^d个corners)
#    读取每个first-stage变量的lb/ub，构造所有组合
# ============================================================
def generate_corners(fsv):
    lows  = []
    highs = []
    for v in fsv:
        lo = v.lb
        hi = v.ub
        if lo is None or hi is None:
            raise ValueError(
                f"Variable {v.name} is missing finite bounds: lb={lo}, ub={hi}"
            )
        if lo > hi:
            raise ValueError(
                f"Variable {v.name} has lb > ub: lb={lo}, ub={hi}"
            )
        lows.append(float(lo))
        highs.append(float(hi))

    lows  = np.array(lows, dtype=float)
    highs = np.array(highs, dtype=float)

    d = len(fsv)
    if d == 0:
        raise ValueError("No first-stage variables in fsv.")

    nodes = []
    for mask in range(1 << d):  # 遍历 0..2^d-1
        point = []
        for i in range(d):
            if (mask >> i) & 1:
                point.append(highs[i])
            else:
                point.append(lows[i])
        nodes.append(np.array(point, dtype=float))
    return nodes


# ============================================================
# 4. 评估角点的目标值(期望)
#    将来可以换成二阶段求解+加权，这里直接用f_true_numpy
# ============================================================
def evaluate_objective_at(node):
    return f_true_numpy(node)


# ============================================================
# 5. 在每个三角形里解 ms, LB, UB
#    注意所有值都是"期望目标"
# ============================================================
def minimize_on_each_triangle(nodes, values, solver='ipopt', tee=False):
    """
    nodes: list of R^2 points
    values: list of f_true_numpy(node) (期望值)
    返回:
      tri: Delaunay剖分
      results: 每个三角形的ms/LB/UB等信息
    """
    pts  = np.asarray(nodes, dtype=float)   # (N,2)
    vals = np.asarray(values, dtype=float)  # (N,)
    tri = Delaunay(pts)                     # tri.simplices: (K,3)

    # 我们用一个"基模型"的结构，然后在循环里clone，避免污染
    m_base, _ = build_model()
    opt = pyo.SolverFactory(solver)
    if solver == "gurobi":
        opt.options["NonConvex"] = 2

    results = []
    for k, simp in enumerate(tri.simplices):
        verts  = pts[simp]     # (3,2)
        fverts = vals[simp]    # (3,)

        # clone独立模型，避免累积约束冲突
        m = m_base.clone()

        # barycentric λ_j
        m.lam = pyo.Var(range(3), domain=pyo.NonNegativeReals)
        m.lam_sum = pyo.Constraint(expr=sum(m.lam[j] for j in range(3)) == 1.0)

        # x = Σ λ_j * v_jx,  y = Σ λ_j * v_jy
        m.x_link_x = pyo.Constraint(
            expr=m.x == sum(m.lam[j]*float(verts[j,0]) for j in range(3))
        )
        m.x_link_y = pyo.Constraint(
            expr=m.y == sum(m.lam[j]*float(verts[j,1]) for j in range(3))
        )

        # As = Σ λ_j * f(v_j)  (线性插值的代理值，对应期望值的线性插值)
        m.As = pyo.Var()
        m.As_con = pyo.Constraint(
            expr=m.As == sum(m.lam[j]*float(fverts[j]) for j in range(3))
        )

        # 目标: min [ E[f(x,y)] - As ]
        m.obj = pyo.Objective(expr=f_true_expr(m) - m.As, sense=pyo.minimize)

        res = opt.solve(m, tee=tee)

        x_ms = np.array([pyo.value(m.x), pyo.value(m.y)], dtype=float)
        ms_val = float(pyo.value(m.obj))       # 该三角形的 ms
        f_at_x_ms = f_true_numpy(x_ms)         # 期望函数在 x_ms 的真实值

        # LB = min As + ms = min f(v) + ms
        LB = float(np.min(fverts) + ms_val)

        # UB = min( min f(v), f(x_ms) )
        UB = float(min(np.min(fverts), f_at_x_ms))

        results.append({
            "simplex_index": int(k),
            "vertices_index": simp.tolist(),
            "verts": verts,
            "fverts": fverts,
            "x_ms": x_ms,
            "ms": ms_val,
            "LB": LB,
            "UB": UB,
            "f(x_ms)": f_at_x_ms,
            "status": str(res.solver.status),
            "term": str(res.solver.termination_condition),
        })

    return tri, results


# ============================================================
# 6. 全局LB的口径：v*是当前nodes中期望值最小的点
#    LB_iter = f(v*) + max_{Δ包含v*} ms_Δ
# ============================================================
def compute_lb_min_node_max_ms(nodes, values, per_tri):
    idx_min = int(np.argmin(values))
    ms_candidates = [r['ms'] for r in per_tri if idx_min in r['vertices_index']]
    if ms_candidates:
        return float(values[idx_min] + max(ms_candidates)), idx_min, float(max(ms_candidates))
    # fallback（几乎不会发生，除非退化三角形等）
    fallback = float(min(r['LB'] for r in per_tri))
    return fallback, idx_min, float('nan')


# ============================================================
# 7. 选点策略（和你原来的一样）
# ============================================================
def pick_candidate_by_ms_rank(per_tri, nodes, min_dist=1e-6, descending=False):
    """
    排序 per_tri: 默认 ms 从小到大 (descending=False)
    然后找一个 x_ms 距离现有节点都 >= min_dist 的
    """
    cands = sorted(per_tri, key=lambda r: r['ms'], reverse=descending)

    def ok(p):
        return all(np.linalg.norm(p - q) >= min_dist for q in nodes)

    for rank, r in enumerate(cands, start=1):
        p = np.array(r['x_ms'], float)
        if ok(p):
            return p, rank, float(r['ms'])

    return None, None, None


def print_per_triangle_table(results, active_mask):
    results = sorted(results, key=lambda r: r["simplex_index"])
    simplex_ids = [r["simplex_index"] for r in results]
    active_set = {sid for sid in simplex_ids if active_mask[sid]}

    header = ["row\\simp"] + [f"T{sid}{'*' if sid in active_set else ''}" for sid in simplex_ids]
    rows = [
        ["UB"] + [f"{r['UB']:.6f}" for r in results],
        ["LB"] + [f"{r['LB']:.6f}" for r in results],
        ["ms"] + [f"{r['ms']:.3e}" for r in results],
    ]
    table = [header] + rows

    ncols = len(header)
    colw = [0]*ncols
    for c in range(ncols):
        colw[c] = max(len(str(row[c])) for row in table) + 2

    RED = "\033[31m"
    RESET = "\033[0m"

    def colorize_if_active(col_idx: int, s: str) -> str:
        if col_idx == 0:
            return s
        sid = simplex_ids[col_idx-1]
        return f"{RED}{s}{RESET}" if sid in active_set else s

    header_line = "".join(
        str(header[c]).ljust(colw[c]) if c == 0
        else colorize_if_active(c, str(header[c]).ljust(colw[c]))
        for c in range(ncols)
    )

    print("\n== Per-triangle summary ==")
    print(header_line)
    print("-" * sum(colw))

    for r in rows:
        line = []
        for c in range(ncols):
            cell = str(r[c])
            padded = cell.ljust(colw[c]) if c == 0 else cell.rjust(colw[c])
            line.append(colorize_if_active(c, padded))
        print("".join(line))

    print("(红色列 = active；第1行=UB，第2行=LB，第3行=ms)\n")


# ============================================================
# 8. 每一轮的3D可视化 (期望曲面 + active单形屋顶)
# ============================================================
def plot_iteration(iter_id, nodes, tri, active_mask,
                   grid_n=60, elev=35, azim=-135,
                   min_node=None, new_node=None,
                   per_tri=None):
    nodes = np.asarray(nodes, float)

    COLOR_SURFACE = 'C0'           # 真实期望函数曲面
    COLOR_NODE_EDGES = '#444444'   # active 单形在真实曲面上的边
    COLOR_PIECE = '#ff7043'        # As+ms 屋顶
    LW_MAIN = 2.2

    # 画期望函数曲面
    # 注意：我们这里用的是 build_model 里的 bounds，先取[0,10]×[0,10]固定网格
    xs = np.linspace(0, 10, grid_n)
    ys = np.linspace(0, 10, grid_n)
    XX, YY = np.meshgrid(xs, ys)
    ZZ = np.zeros_like(XX)
    for i in range(XX.shape[0]):
        for j in range(XX.shape[1]):
            ZZ[i,j] = f_true_numpy([XX[i,j], YY[i,j]])

    fig = plt.figure(figsize=(8,6))
    ax = fig.add_subplot(111, projection='3d')

    ax.plot_surface(XX, YY, ZZ, alpha=0.35, linewidth=0,
                    antialiased=True, color=COLOR_SURFACE)

    # active单形的 As+ms 屋顶
    if per_tri is not None and len(per_tri) > 0 and tri is not None:
        faces = []
        for r in per_tri:
            sid = r["simplex_index"]
            if not active_mask.get(sid, False):
                continue
            verts_xy = r["verts"]     # (3,2)
            fverts   = r["fverts"]    # (3,)
            ms_val   = r["ms"]
            tri_xyz  = [
                (verts_xy[i,0], verts_xy[i,1], float(fverts[i] + ms_val))
                for i in range(3)
            ]
            faces.append(tri_xyz)

        if faces:
            coll = Poly3DCollection(
                faces,
                facecolors=[COLOR_PIECE],
                edgecolors='none',
                alpha=0.55
            )
            ax.add_collection3d(coll)

        # active单形屋顶的边线
        for r in per_tri:
            sid = r["simplex_index"]
            if not active_mask.get(sid, False):
                continue
            verts_xy = r["verts"]
            fverts   = r["fverts"]
            ms_val   = r["ms"]
            for i in range(3):
                a_xy = verts_xy[i]
                b_xy = verts_xy[(i+1) % 3]
                za = float(fverts[i] + ms_val)
                zb = float(fverts[(i+1) % 3] + ms_val)
                ax.plot([a_xy[0], b_xy[0]],
                        [a_xy[1], b_xy[1]],
                        [za, zb],
                        lw=LW_MAIN, c=COLOR_PIECE)

    # 所有节点 (黑点)
    node_z = np.array([f_true_numpy(p) for p in nodes])
    ax.scatter(nodes[:,0], nodes[:,1], node_z,
               s=30, c='k', depthshade=False)

    # active 单形在真实曲面上的边 + 质心
    if tri is not None:
        for k, simp in enumerate(tri.simplices):
            if not active_mask.get(k, False):
                continue
            verts_xy = tri.points[simp]
            for i in range(3):
                a, b = verts_xy[i], verts_xy[(i+1) % 3]
                ax.plot([a[0], b[0]], [a[1], b[1]],
                        [f_true_numpy(a), f_true_numpy(b)],
                        lw=LW_MAIN, c=COLOR_NODE_EDGES)

            centroid = np.mean(verts_xy, axis=0)
            ax.scatter([centroid[0]], [centroid[1]],
                       [f_true_numpy(centroid)],
                       marker='^', c=COLOR_NODE_EDGES,
                       s=70, depthshade=False)

    # 当前最优节点 (绿色)
    if min_node is not None:
        ax.scatter([min_node[0]], [min_node[1]],
                   [f_true_numpy(min_node)],
                   color='g', s=80, marker='o', label='Current min node')

    # 下一轮即将加入的新节点 (蓝色)
    if new_node is not None:
        ax.scatter([new_node[0]], [new_node[1]],
                   [f_true_numpy(new_node)],
                   color='b', s=80, marker='o', label='Next node')

    ax.set_title(f"Iteration {iter_id}")
    ax.set_xlabel("x")
    ax.set_ylabel("y")
    ax.set_zlabel("E[f(x,y)]")
    ax.legend(loc='upper left')
    ax.view_init(elev=elev, azim=azim)
    plt.tight_layout()
    plt.show()


# ============================================================
# 9. 收敛曲线 (LB/UB/ms vs node_count)
# ============================================================
def plot_convergence(hist, prec=4):
    x  = hist["node_count"]
    LB = hist["LB_hist"]
    UB = hist["UB_hist"]
    ms = hist["ms_hist"]

    # LB / UB
    plt.figure(figsize=(6,4))
    plt.plot(x, LB, marker='o', label="LB (lower bound)")
    plt.plot(x, UB, marker='s', label="UB (upper bound)")
    plt.xlabel("Number of nodes")
    plt.ylabel("Bound value")
    plt.title("Global LB and UB vs Number of Nodes")
    plt.legend()
    plt.grid(True)
    plt.tight_layout()

    # ms
    plt.figure(figsize=(6,4))
    plt.plot(x, ms, marker='o', color='C2')
    plt.xlabel("Number of nodes")
    plt.ylabel("Global ms (min over active simplices)")
    plt.title("ms vs Number of Nodes (Simplex method)")
    plt.grid(True)
    plt.tight_layout()
    plt.show()

    # 表格
    print("\n==== Summary Table ====")
    fmt = f"{{:.{prec}f}}"
    header = f"{'Nodes':>8s} | {'ms':>12s} | {'LB':>12s} | {'UB':>12s}"
    print(header)
    print("-" * len(header))
    for n, m_val, lb, ub in zip(x, ms, LB, UB):
        print(f"{n:8d} | {fmt.format(m_val):>12s} | {fmt.format(lb):>12s} | {fmt.format(ub):>12s}")
    print("-" * len(header))
    print(f"{'Total':>8s} | {len(x):>12d}")


# ============================================================
# 10. 主循环 run_case
#     和单场景改造后一样：不再自己造初始点
#     吃 init_nodes / init_values
# ============================================================
def run_case(init_nodes,
             init_values,
             max_nodes=10,
             solver='ipopt',
             tee=False,
             plot_grid_n=60,
             record_history=True,
             make_3d_plots=True,
             only_from_active=True,
             min_dist=1e-4,
             active_tol=1e-8,
             print_tables=True):
    """
    init_nodes: generate_corners(fsv) 的输出
    init_values: [ evaluate_objective_at(node) for node in init_nodes ]
    其他参数同以前
    """
    # 拷贝，避免原对象被原地修改
    nodes  = [np.array(p, dtype=float) for p in init_nodes]
    values = [float(v) for v in init_values]

    LB_hist, UB_hist, ms_hist, node_count = [], [], [], []

    it = 0
    while len(nodes) < max_nodes:
        tri, per_tri = minimize_on_each_triangle(
            nodes, values, solver=solver, tee=tee
        )

        # UB_global = 当前节点集中期望目标的最小值
        UB_global = float(np.min(values))

        # LB_iter = v*的值 + 其相邻单形 ms 的最大值
        LB_iter, idx_min_node, ms_chosen = compute_lb_min_node_max_ms(
            nodes, values, per_tri
        )

        # ms_iter = 所有单形里的最小 ms
        ms_iter = min(r['ms'] for r in per_tri)

        # 历史记录
        if record_history:
            LB_hist.append(LB_iter)
            UB_hist.append(UB_global)
            ms_hist.append(ms_iter)
            node_count.append(len(nodes))

        # active 单形
        active_mask = {
            r['simplex_index']: (r['LB'] <= UB_global + active_tol)
            for r in per_tri
        }
        active_tris = [r for r in per_tri if r['LB'] <= UB_global + active_tol]

        # 选择候选集合
        candidates = active_tris if (only_from_active and len(active_tris) > 0) else per_tri
        if only_from_active and len(active_tris) == 0:
            print(f"[warn] No active simplices this round (tol={active_tol:g}); "
                  f"fall back to all simplices.")

        # 选新节点：按 ms 最小优先，同时保持最小距离
        new_node, used_rank, used_ms = pick_candidate_by_ms_rank(
            candidates, nodes,
            min_dist=min_dist,
            descending=False
        )
        if new_node is None:
            print("All candidates too close; stop.")
            break

        # 输出本轮表格
        if print_tables:
            print(f"\n===== Iteration {it} =====")
            print(f"UB_global = {UB_global:.6f}")
            print(f"LB_iter   = {LB_iter:.6f}")
            print(f"ms_iter   = {ms_iter:.6e}")
            print_per_triangle_table(per_tri, active_mask)

        # 画3D (只有2D的一阶段变量我们才能画这个曲面图)
        if make_3d_plots and len(nodes[0]) == 2:
            plot_iteration(
                it, nodes, tri, active_mask,
                grid_n=plot_grid_n,
                min_node=nodes[idx_min_node],
                new_node=new_node,
                per_tri=per_tri
            )

        # 把新节点加入，如果和已有点太近就停止
        if any(np.linalg.norm(new_node - q) < min_dist for q in nodes):
            print("New node too close to an existing node; stop.")
            break

        nodes.append(new_node)
        values.append(evaluate_objective_at(new_node))

        it += 1

    return {
        "nodes": np.array(nodes),
        "values": np.array(values),
        "LB_hist": LB_hist,
        "UB_hist": UB_hist,
        "ms_hist": ms_hist,
        "node_count": node_count
    }


# ============================================================
# 11. 随机生成场景 (和你原来一样)
# ============================================================
def random_scenarios(num_scenarios,
                     domain_low=0.0, domain_high=10.0,
                     depth_range=(0.8, 2.0),
                     width_range=(0.8, 2.0),
                     bias_strength_range=(0.02, 0.08),
                     bias_center=(5.0,5.0),
                     offset=0.0,
                     seed=None):
    rng = np.random.default_rng(seed)
    scens = []
    for _ in range(num_scenarios):
        cx = rng.uniform(domain_low, domain_high)
        cy = rng.uniform(domain_low, domain_high)
        depth = rng.uniform(*depth_range)
        width = rng.uniform(*width_range)
        bs = rng.uniform(*bias_strength_range)
        scens.append({
            "center": np.array([cx, cy], dtype=float),
            "depth": float(depth),
            "width": float(width),
            "outer_bias_strength": float(bs),
            "outer_bias_center": np.array(bias_center, dtype=float),
            "offset": float(offset),
            # p 之后统一设置
        })
    # 均匀权重
    pval = 1.0/len(scens)
    for s in scens:
        s["p"] = pval
    return scens


# ============================================================
# 12. main：生成场景 -> 生成初始角点 -> 跑算法 -> 画收敛
# ============================================================
if __name__ == "__main__":
    # 1) 随机生成多场景
    SCENARIOS = random_scenarios(
        num_scenarios=3,
        seed=42  # 固定随机数种子以复现
    )

    # 2) 拿到一个模型 + fsv
    model_for_bounds, fsv = build_model()

    # 3) 用 fsv 的 bounds 自动生成所有角点
    init_nodes = generate_corners(fsv)

    # 4) 评估这些角点的期望目标值
    init_values = [evaluate_objective_at(p) for p in init_nodes]

    # 5) 运行主循环
    solver      = "ipopt"   # 也可以"gurobi"
    max_nodes   = 20
    plot_3d     = True
    record_hist = True

    hist_simp = run_case(
        init_nodes=init_nodes,
        init_values=init_values,
        max_nodes=max_nodes,
        solver=solver,
        tee=False,
        plot_grid_n=60,
        record_history=record_hist,
        make_3d_plots=plot_3d,
        only_from_active=True,
        min_dist=1e-4,
        active_tol=1e-8,
        print_tables=True
    )

    # 6) 收敛曲线
    plot_convergence(hist_simp, prec=4)

    # 7) 最后打印
    nodes  = hist_simp["nodes"]
    values = hist_simp["values"]
    print("\n==== Done ====")
    print(f"Total nodes: {len(nodes)}")
    print(f"Best expected f among nodes: {np.min(values):.6f}")
```
